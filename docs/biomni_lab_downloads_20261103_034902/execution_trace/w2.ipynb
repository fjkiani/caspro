````python

import os, textwrap

REPO = "/mnt/shared-workspace/shared/crispro-backend-v3-base"

def write(path, content):
    full = os.path.join(REPO, path)
    os.makedirs(os.path.dirname(full), exist_ok=True)
    with open(full, "w") as f:
        f.write(textwrap.dedent(content))
    print(f"  wrote {path} ({os.path.getsize(full)} bytes)")

# ── LAYER 1: sl_therapy_bridge/__init__.py ────────────────────────────────────
write("layer1_engines/sl_therapy_bridge/__init__.py", '''
    """SL Therapy Bridge — ALLOW/WARN/DISALLOW policy engine."""
    from .bridge import SLTherapyBridge, evaluate_sl_therapy_bridge
    from .models import SLBridgeInput, SLBridgeResult, TherapyRecommendation, PolicyDecision
    from .policy_engine import SL_THERAPY_MAP, PolicyEngine

    __all__ = [
        "SLTherapyBridge", "evaluate_sl_therapy_bridge",
        "SLBridgeInput", "SLBridgeResult", "TherapyRecommendation", "PolicyDecision",
        "SL_THERAPY_MAP", "PolicyEngine",
    ]
    ''')

# ── LAYER 1: sl_therapy_bridge/models.py ─────────────────────────────────────
write("layer1_engines/sl_therapy_bridge/models.py", '''
    """
    SL Therapy Bridge — Data Models
    RUO — Research Use Only.
    """

    from dataclasses import dataclass, field
    from typing import List, Optional
    from enum import Enum


    class PolicyDecision(str, Enum):
        ALLOW = "ALLOW"
        ALLOW_PRECLINICAL_ONLY = "ALLOW_PRECLINICAL_ONLY"
        WARN = "WARN"
        DISALLOW = "DISALLOW"
        NOT_IN_MAP = "NOT_IN_MAP"


    @dataclass
    class TherapyRecommendation:
        sl_partner: str
        driver: str
        therapy: str
        decision: PolicyDecision
        rationale: str
        evidence_tier: str
        mandatory_disclosure: Optional[str] = None


    @dataclass
    class SLBridgeInput:
        patient_id: str
        sl_partners: List[str]          # e.g. ["ZEB1", "BRCA2_mut", "ATM_mut"]
        candidate_therapies: List[str]  # e.g. ["olaparib", "ceralasertib"]


    @dataclass
    class SLBridgeResult:
        patient_id: str
        recommendations: List[TherapyRecommendation]
        allow_count: int
        warn_count: int
        disallow_count: int
        not_in_map_count: int
        top_recommendation: Optional[TherapyRecommendation]
        ruo: bool = True
        disclaimer: str = "Research Use Only. Not cleared for clinical decision-making."
    ''')

# ── LAYER 1: sl_therapy_bridge/policy_engine.py ───────────────────────────────
write("layer1_engines/sl_therapy_bridge/policy_engine.py", '''
    """
    SL Therapy Bridge — Policy Engine

    ALLOW/WARN/DISALLOW logic for SL partner → therapy mapping.
    All decisions sourced from KG v2 verified anchors.

    CRITICAL EXCLUSION:
      BRIEF-2 (PTEN-null → ITGAV) is a COMPUTATIONAL HYPOTHESIS.
      DepMap NULL across all 3 proxies. MUST NOT appear in this map.
      Mandatory disclosure required in every context where BRIEF-2 is mentioned.

    RUO — Research Use Only.
    """

    from typing import Dict, List, Optional, Tuple
    from .models import PolicyDecision, TherapyRecommendation


    # SL Therapy Map — ALLOW/WARN/DISALLOW
    # Key: (sl_partner_normalized, therapy_normalized)
    # Value: (decision, rationale, evidence_tier, mandatory_disclosure)
    SL_THERAPY_MAP: Dict[Tuple[str, str], Tuple[PolicyDecision, str, str, Optional[str]]] = {
        # ── BRCA2_mut ──────────────────────────────────────────────────────────
        ("brca2_mut", "olaparib"): (
            PolicyDecision.ALLOW,
            "FDA-approved: BRCA1/2-mutated HER2-negative breast cancer, BRCA1/2-mutated mCRPC, "
            "BRCA1/2-mutated pancreatic cancer. Synthetic lethal with BRCA2 loss.",
            "FDA_APPROVED",
            None,
        ),
        ("brca2_mut", "niraparib"): (
            PolicyDecision.ALLOW,
            "FDA-approved: BRCA1/2-mutated ovarian cancer (maintenance). Synthetic lethal with BRCA2 loss.",
            "FDA_APPROVED",
            None,
        ),
        ("brca2_mut", "rucaparib"): (
            PolicyDecision.ALLOW,
            "FDA-approved: BRCA1/2-mutated ovarian cancer. Synthetic lethal with BRCA2 loss.",
            "FDA_APPROVED",
            None,
        ),
        # ── BRCA1_mut ──────────────────────────────────────────────────────────
        ("brca1_mut", "olaparib"): (
            PolicyDecision.ALLOW,
            "FDA-approved: BRCA1/2-mutated HER2-negative breast cancer. Synthetic lethal with BRCA1 loss.",
            "FDA_APPROVED",
            None,
        ),
        ("brca1_mut", "niraparib"): (
            PolicyDecision.ALLOW,
            "FDA-approved: BRCA1/2-mutated ovarian cancer (maintenance). Synthetic lethal with BRCA1 loss.",
            "FDA_APPROVED",
            None,
        ),
        ("brca1_mut", "rucaparib"): (
            PolicyDecision.ALLOW,
            "FDA-approved: BRCA1/2-mutated ovarian cancer. Synthetic lethal with BRCA1 loss.",
            "FDA_APPROVED",
            None,
        ),
        # ── ATM_mut ────────────────────────────────────────────────────────────
        ("atm_mut", "ceralasertib"): (
            PolicyDecision.ALLOW,
            "ATM-mutated tumors are hypersensitive to ATR inhibition (ceralasertib). "
            "RS-High ORR 40% vs RS-Low 5% (KG v2 ATR/DDR corpus). Clinical-stage.",
            "CLINICAL_STAGE",
            None,
        ),
        ("atm_mut", "berzosertib"): (
            PolicyDecision.ALLOW,
            "ATM-mutated tumors are hypersensitive to ATR inhibition (berzosertib). "
            "RS-High ORR 40% vs RS-Low 5% (KG v2 ATR/DDR corpus). Clinical-stage.",
            "CLINICAL_STAGE",
            None,
        ),
        # ── ARID1A_mut ─────────────────────────────────────────────────────────
        ("arid1a_mut", "tazemetostat"): (
            PolicyDecision.ALLOW,
            "ARID1A-mutated tumors are synthetic lethal with EZH2 inhibition (tazemetostat). "
            "FDA-approved for EZH2-mutated follicular lymphoma; clinical evidence in solid tumors.",
            "CLINICAL_STAGE",
            None,
        ),
        # ── SMARCA4_mut ────────────────────────────────────────────────────────
        ("smarca4_mut", "fhd-909"): (
            PolicyDecision.ALLOW_PRECLINICAL_ONLY,
            "SMARCA4-mutated tumors are synthetic lethal with SMARCA2 inhibition (FHD-909). "
            "Preclinical data only — not yet in clinical trials. ALLOW_PRECLINICAL_ONLY.",
            "PRECLINICAL",
            "FHD-909 is in preclinical development only. Not cleared for clinical use.",
        ),
        # ── CCNE1_AMP ──────────────────────────────────────────────────────────
        ("ccne1_amp", "cdk2_inhibitor"): (
            PolicyDecision.ALLOW,
            "CCNE1 amplification drives CDK2 dependency. CDK2 inhibitors are synthetic lethal "
            "with CCNE1 amplification. Multiple CDK2 inhibitors in clinical trials.",
            "CLINICAL_STAGE",
            None,
        ),
        # ── KRAS_mut — DISALLOW ────────────────────────────────────────────────
        ("kras_mut", "egfr_tki"): (
            PolicyDecision.DISALLOW,
            "KRAS mutation confers primary resistance to EGFR TKIs. EGFR TKIs are "
            "contraindicated in KRAS-mutated CRC (FDA label). Mechanism: KRAS activates "
            "downstream signaling independently of EGFR.",
            "FDA_CONTRAINDICATED",
            None,
        ),
        # ── PTEN_null — WARN ───────────────────────────────────────────────────
        ("pten_null", "mtor_inhibitor"): (
            PolicyDecision.WARN,
            "PTEN loss activates PI3K/mTOR pathway — mTOR inhibitors are mechanistically "
            "rational but clinical evidence is mixed. Adavosertib: PTEN-intact ORR 23% vs "
            "PTEN-loss ORR 0%; PTEN-loss PFS HR=1.82 (HARMFUL). Proceed with caution.",
            "CLINICAL_MIXED",
            "PTEN-loss patients showed HARMFUL outcomes with WEE1 inhibition (HR=1.82). "
            "mTOR inhibitor benefit in PTEN-null is not established in CRC.",
        ),
        ("pten_null", "pi3k_inhibitor"): (
            PolicyDecision.WARN,
            "PTEN loss activates PI3K pathway — PI3K inhibitors are mechanistically rational "
            "but clinical evidence is limited in CRC. Proceed with caution.",
            "CLINICAL_LIMITED",
            "PI3K inhibitor benefit in PTEN-null CRC is not established.",
        ),
        # ── ZEB1 — BRIEF-1 (DEAL_READY) ───────────────────────────────────────
        ("zeb1", "itgav_inhibitor"): (
            PolicyDecision.ALLOW,
            "BRIEF-1: ZEB1→ITGAV synthetic lethality. delta=−0.7184, FDR=0.001203, "
            "CRISPR_VALIDATED, DEAL_READY. ZEB1 high → ITGAV essential for survival. "
            "Mechanistic basis: ZEB1-driven EMT increases integrin αV dependency.",
            "CRISPR_VALIDATED",
            None,
        ),
        # ── SPP1 — BRIEF-4 ─────────────────────────────────────────────────────
        ("spp1", "nrf2_inhibitor"): (
            PolicyDecision.ALLOW,
            "BRIEF-4: SPP1→NRF2 synthetic lethality. delta=−0.7326, FDR=8×10⁻⁶, "
            "CRISPR_VALIDATED. SPP1 high → NRF2 essential for oxidative stress defense.",
            "CRISPR_VALIDATED",
            None,
        ),
        # ── PTEN_null → ITGAV — BRIEF-2 — EXPLICITLY EXCLUDED ─────────────────
        # BRIEF-2 (PTEN-null → ITGAV) is a COMPUTATIONAL HYPOTHESIS.
        # DepMap NULL across all 3 proxies. MUST NOT be in this map.
        # If queried, the bridge returns NOT_IN_MAP with mandatory disclosure.
    }


    class PolicyEngine:
        """Evaluates SL partner + therapy pairs against the policy map."""

        def evaluate_pair(
            self, sl_partner: str, therapy: str
        ) -> TherapyRecommendation:
            key = (sl_partner.lower().replace("-", "_"), therapy.lower().replace("-", "_"))

            # Special case: BRIEF-2 PTEN-null → ITGAV — mandatory disclosure
            if key[0] in ("pten_null", "pten-null") and key[1] in ("itgav", "itgav_inhibitor"):
                return TherapyRecommendation(
                    sl_partner=sl_partner,
                    driver="PTEN_null",
                    therapy=therapy,
                    decision=PolicyDecision.NOT_IN_MAP,
                    rationale=(
                        "BRIEF-2 (PTEN-null → ITGAV) is a COMPUTATIONAL HYPOTHESIS. "
                        "DepMap NULL across all 3 proxies. Not included in the SL therapy map."
                    ),
                    evidence_tier="COMPUTATIONAL_HYPOTHESIS",
                    mandatory_disclosure=(
                        "PTEN-null → ITGAV is a COMPUTATIONAL HYPOTHESIS. "
                        "DepMap NULL across all 3 proxies. Not validated by CRISPR."
                    ),
                )

            if key in SL_THERAPY_MAP:
                decision, rationale, tier, disclosure = SL_THERAPY_MAP[key]
                return TherapyRecommendation(
                    sl_partner=sl_partner,
                    driver=sl_partner,
                    therapy=therapy,
                    decision=decision,
                    rationale=rationale,
                    evidence_tier=tier,
                    mandatory_disclosure=disclosure,
                )

            return TherapyRecommendation(
                sl_partner=sl_partner,
                driver=sl_partner,
                therapy=therapy,
                decision=PolicyDecision.NOT_IN_MAP,
                rationale=f"No policy entry for ({sl_partner}, {therapy}) — not in SL therapy map.",
                evidence_tier="UNKNOWN",
                mandatory_disclosure=None,
            )
    ''')

# ── LAYER 1: sl_therapy_bridge/bridge.py ─────────────────────────────────────
write("layer1_engines/sl_therapy_bridge/bridge.py", '''
    """
    SL Therapy Bridge — Main Bridge Class
    RUO — Research Use Only.
    """

    from typing import Optional
    from .models import SLBridgeInput, SLBridgeResult, TherapyRecommendation, PolicyDecision
    from .policy_engine import PolicyEngine


    class SLTherapyBridge:
        """
        Bridges SL partner findings to therapy recommendations.
        Runs as an additive post-processor after the SL engine (Phase 5a).
        """

        def __init__(self):
            self._engine = PolicyEngine()

        def evaluate(self, inp: SLBridgeInput) -> SLBridgeResult:
            recommendations = []
            for partner in inp.sl_partners:
                for therapy in inp.candidate_therapies:
                    rec = self._engine.evaluate_pair(partner, therapy)
                    recommendations.append(rec)

            allow = sum(1 for r in recommendations if r.decision == PolicyDecision.ALLOW)
            warn = sum(1 for r in recommendations if r.decision == PolicyDecision.WARN)
            disallow = sum(1 for r in recommendations if r.decision == PolicyDecision.DISALLOW)
            not_in_map = sum(1 for r in recommendations if r.decision in (
                PolicyDecision.NOT_IN_MAP, PolicyDecision.ALLOW_PRECLINICAL_ONLY
            ))

            top = self._top_recommendation(recommendations)

            return SLBridgeResult(
                patient_id=inp.patient_id,
                recommendations=recommendations,
                allow_count=allow,
                warn_count=warn,
                disallow_count=disallow,
                not_in_map_count=not_in_map,
                top_recommendation=top,
            )

        def _top_recommendation(
            self, recs: list
        ) -> Optional[TherapyRecommendation]:
            priority = {
                PolicyDecision.ALLOW: 0,
                PolicyDecision.ALLOW_PRECLINICAL_ONLY: 1,
                PolicyDecision.WARN: 2,
                PolicyDecision.DISALLOW: 3,
                PolicyDecision.NOT_IN_MAP: 4,
            }
            allowed = [r for r in recs if r.decision == PolicyDecision.ALLOW]
            if allowed:
                # Prefer CRISPR_VALIDATED > FDA_APPROVED > CLINICAL_STAGE
                tier_priority = {"CRISPR_VALIDATED": 0, "FDA_APPROVED": 1, "CLINICAL_STAGE": 2}
                return sorted(allowed, key=lambda r: tier_priority.get(r.evidence_tier, 99))[0]
            return sorted(recs, key=lambda r: priority.get(r.decision, 99))[0] if recs else None


    def evaluate_sl_therapy_bridge(inp: SLBridgeInput) -> SLBridgeResult:
        """Convenience function."""
        return SLTherapyBridge().evaluate(inp)
    ''')

# ── LAYER 1: ceacam5_biomarker_strategy/__init__.py ──────────────────────────
write("layer1_engines/ceacam5_biomarker_strategy/__init__.py", '''
    """CEACAM5 Biomarker Strategy Engine — Two-Gate Framework."""
    from .engine import Ceacam5BiomarkerEngine, evaluate_ceacam5_strategy
    from .models import Ceacam5Input, Ceacam5Result, GateStatus
    from .vector_scalpel import VectorScalpel, VECTOR_SCALPEL_RANKING

    __all__ = [
        "Ceacam5BiomarkerEngine", "evaluate_ceacam5_strategy",
        "Ceacam5Input", "Ceacam5Result", "GateStatus",
        "VectorScalpel", "VECTOR_SCALPEL_RANKING",
    ]
    ''')

# ── LAYER 1: ceacam5_biomarker_strategy/models.py ────────────────────────────
write("layer1_engines/ceacam5_biomarker_strategy/models.py", '''
    """
    CEACAM5 Biomarker Strategy Engine — Data Models

    Two-gate framework for CEACAM5-targeted therapy selection.
    Evidence sources: KG v2 CEACAM5 corpus (Jansen 2024, Warmerdam 2025,
    CARMEN-LC03, PROCEADE-CRC-01, CO.26 biomarker analysis).
    RUO — Research Use Only.
    """

    from dataclasses import dataclass, field
    from typing import List, Optional
    from enum import Enum


    class GateStatus(str, Enum):
        PASS = "PASS"
        FAIL = "FAIL"
        UNKNOWN = "UNKNOWN"


    @dataclass
    class Ceacam5Input:
        patient_id: str
        disease: str                            # e.g. "MSS mCRC", "NSCLC"
        # Gate 1: CEACAM5 IHC ≥80% (primary gate)
        ceacam5_ihc_pct: Optional[float] = None  # 0–100; None = not tested
        # Gate 2: cCEA ≥100 ng/mL (surrogate gate)
        ccea_ngml: Optional[float] = None        # None = not collected
        # Additional biomarkers
        ptmb: Optional[float] = None             # mut/Mb
        immunoscore_ic: Optional[str] = None     # "HIGH" / "LOW"
        liver_metastasis: Optional[bool] = None


    @dataclass
    class Ceacam5Result:
        patient_id: str
        gate_1_status: GateStatus
        gate_1_evidence: str
        gate_2_status: GateStatus
        gate_2_evidence: str
        overall_status: GateStatus
        overall_rationale: str
        vector_scalpel_ranking: List[str]
        open_assumption_oa08: str
        recommended_biomarker_panel: List[str]
        evidence_anchors: List[str]
        ruo: bool = True
        disclaimer: str = "Research Use Only. Not cleared for clinical decision-making."
    ''')

# ── LAYER 1: ceacam5_biomarker_strategy/vector_scalpel.py ────────────────────
write("layer1_engines/ceacam5_biomarker_strategy/vector_scalpel.py", '''
    """
    Vector-Scalpel Biomarker Ranking for CEACAM5 Strategy

    Ranking of biomarkers by predictive power for CEACAM5-targeted therapy
    in MSS CRC. Source: KG v2 CEACAM5 corpus (Vector-Scalpel analysis).

    Ranking (out of 33 patients):
      1. cCEA (27/33) — strongest surrogate for CEACAM5 IHC ≥80%
      2. pTMB (23/33) — IO enrichment signal
      3. IHC (20/33) — direct CEACAM5 expression
      4. Immunoscore (19/33) — TME inflammation
      5. DetermaIO (15/33) — IO signature

    Note: IHC ↔ cCEA Spearman ρ = 0.43 (n=92) — moderate correlation.
    cCEA ≥100 ORR in CEACAM5 HE NSCLC: 41.7% vs 8.1%, P=0.003.

    RUO — Research Use Only.
    """

    from dataclasses import dataclass
    from typing import List


    @dataclass(frozen=True)
    class BiomarkerRankEntry:
        rank: int
        name: str
        score: str          # e.g. "27/33"
        score_fraction: float
        rationale: str
        threshold: str


    VECTOR_SCALPEL_RANKING: List[BiomarkerRankEntry] = [
        BiomarkerRankEntry(
            rank=1,
            name="cCEA",
            score="27/33",
            score_fraction=27/33,
            rationale=(
                "Strongest surrogate for CEACAM5 IHC ≥80%. "
                "cCEA ≥100 ORR in CEACAM5 HE NSCLC: 41.7% vs 8.1%, P=0.003. "
                "IHC ↔ cCEA Spearman ρ=0.43 (n=92)."
            ),
            threshold="≥100 ng/mL",
        ),
        BiomarkerRankEntry(
            rank=2,
            name="pTMB",
            score="23/33",
            score_fraction=23/33,
            rationale=(
                "IO enrichment signal. CO.26 pTMB≥28 OS HR=0.34 (90% CI 0.18–0.63, p=0.022). "
                "Tissue TMB NOT a substitute (Spearman r=0.13, p=0.20)."
            ),
            threshold="≥28 mut/Mb (GuardantOMNI)",
        ),
        BiomarkerRankEntry(
            rank=3,
            name="IHC",
            score="20/33",
            score_fraction=20/33,
            rationale=(
                "Direct CEACAM5 expression. CARMEN-LC03: IHC 50–79% PFS HR=1.38 (HARMFUL); "
                "IHC ≥80% OS HR=0.71 (post-hoc). PROCEADE-CRC-01: ORR 7.5% (3/40); "
                "IHC Spearman rho=−0.14, P=0.4."
            ),
            threshold="≥80% tumor cells (H-score or % positive)",
        ),
        BiomarkerRankEntry(
            rank=4,
            name="Immunoscore",
            score="19/33",
            score_fraction=19/33,
            rationale=(
                "TME inflammation marker. AtezoTRIBE: Immunoscore-IC high HR ~0.50 (exploratory). "
                "Immunoscore-IC high correlates with IO benefit in MSS CRC."
            ),
            threshold="Immunoscore-IC HIGH",
        ),
        BiomarkerRankEntry(
            rank=5,
            name="DetermaIO",
            score="15/33",
            score_fraction=15/33,
            rationale="IO gene expression signature. Weakest predictor in the CEACAM5 context.",
            threshold="DetermaIO positive",
        ),
    ]


    class VectorScalpel:
        """Ranks biomarkers for CEACAM5 strategy and recommends collection priority."""

        def get_ranked_panel(self) -> List[str]:
            return [
                f"#{e.rank} {e.name} ({e.score}) — threshold: {e.threshold}"
                for e in VECTOR_SCALPEL_RANKING
            ]

        def get_top_n(self, n: int = 3) -> List[BiomarkerRankEntry]:
            return VECTOR_SCALPEL_RANKING[:n]
    ''')

# ── LAYER 1: ceacam5_biomarker_strategy/engine.py ────────────────────────────
write("layer1_engines/ceacam5_biomarker_strategy/engine.py", '''
    """
    CEACAM5 Biomarker Strategy Engine — Two-Gate Framework

    Gate 1: CEACAM5 IHC ≥80% (primary gate — direct expression)
    Gate 2: cCEA ≥100 ng/mL (surrogate gate — serum marker)

    Epidemiology (KG v2 verified):
      - Any CEACAM5 positivity in CRC: 98.7% (Jansen et al. Cancers 2024)
      - CRLM positivity: 79% (Warmerdam et al. EJNMMI Res 2025)
      - IHC ≥80% fraction in MSS mCRC: ~28% (OPEN_ASSUMPTION OA-08)

    OPEN_ASSUMPTION OA-08:
      The 28% IHC ≥80% fraction is an open assumption — not yet validated
      in a prospective MSS mCRC cohort. Must be disclosed in all outputs.

    RUO — Research Use Only.
    """

    from .models import Ceacam5Input, Ceacam5Result, GateStatus
    from .vector_scalpel import VectorScalpel, VECTOR_SCALPEL_RANKING

    # Evidence anchors (KG v2 verified)
    CEACAM5_ANY_POSITIVITY_CRC = 0.987      # Jansen et al. Cancers 2024
    CEACAM5_CRLM_POSITIVITY = 0.79          # Warmerdam et al. EJNMMI Res 2025
    CEACAM5_IHC_GE80_FRACTION = 0.28        # OPEN_ASSUMPTION OA-08
    IHC_THRESHOLD = 80.0                    # % tumor cells
    CCEA_THRESHOLD = 100.0                  # ng/mL
    CARMEN_IHC_50_79_PFS_HR = 1.38          # HARMFUL
    CARMEN_IHC_GE80_OS_HR = 0.71            # post-hoc
    PROCEADE_ORR = 0.075                    # 3/40
    CCEA_ORR_HIGH = 0.417                   # 41.7% in CEACAM5 HE NSCLC
    CCEA_ORR_LOW = 0.081                    # 8.1%

    OPEN_ASSUMPTION_OA08 = (
        "OPEN_ASSUMPTION OA-08: The ~28% IHC ≥80% fraction in MSS mCRC is an open assumption "
        "not yet validated in a prospective cohort. Treat as approximate prevalence estimate only."
    )


    class Ceacam5BiomarkerEngine:
        """Evaluates CEACAM5 biomarker strategy using the two-gate framework."""

        def __init__(self):
            self._scalpel = VectorScalpel()

        def evaluate(self, inp: Ceacam5Input) -> Ceacam5Result:
            g1_status, g1_ev = self._gate1_ihc(inp.ceacam5_ihc_pct)
            g2_status, g2_ev = self._gate2_ccea(inp.ccea_ngml)
            overall, rationale = self._overall(g1_status, g2_status)

            return Ceacam5Result(
                patient_id=inp.patient_id,
                gate_1_status=g1_status,
                gate_1_evidence=g1_ev,
                gate_2_status=g2_status,
                gate_2_evidence=g2_ev,
                overall_status=overall,
                overall_rationale=rationale,
                vector_scalpel_ranking=self._scalpel.get_ranked_panel(),
                open_assumption_oa08=OPEN_ASSUMPTION_OA08,
                recommended_biomarker_panel=self._panel(inp),
                evidence_anchors=[
                    f"CEACAM5 any positivity in CRC: {CEACAM5_ANY_POSITIVITY_CRC:.1%} (Jansen et al. Cancers 2024)",
                    f"CRLM positivity: {CEACAM5_CRLM_POSITIVITY:.0%} (Warmerdam et al. EJNMMI Res 2025)",
                    f"IHC ≥80% fraction in MSS mCRC: ~{CEACAM5_IHC_GE80_FRACTION:.0%} (OPEN_ASSUMPTION OA-08)",
                    f"CARMEN-LC03: IHC 50–79% PFS HR={CARMEN_IHC_50_79_PFS_HR} (HARMFUL); IHC ≥80% OS HR={CARMEN_IHC_GE80_OS_HR} (post-hoc)",
                    f"PROCEADE-CRC-01: ORR {PROCEADE_ORR:.1%} (3/40); IHC Spearman rho=−0.14, P=0.4",
                    f"cCEA ≥100 ORR in CEACAM5 HE NSCLC: {CCEA_ORR_HIGH:.1%} vs {CCEA_ORR_LOW:.1%}, P=0.003",
                ],
            )

        def _gate1_ihc(self, ihc_pct):
            if ihc_pct is None:
                return GateStatus.UNKNOWN, "CEACAM5 IHC not tested — Gate 1 unknown"
            if ihc_pct >= IHC_THRESHOLD:
                return GateStatus.PASS, (
                    f"CEACAM5 IHC={ihc_pct:.0f}% ≥ {IHC_THRESHOLD:.0f}% threshold — Gate 1 PASS. "
                    f"CARMEN-LC03 IHC ≥80% OS HR={CARMEN_IHC_GE80_OS_HR} (post-hoc)."
                )
            if ihc_pct >= 50:
                return GateStatus.FAIL, (
                    f"CEACAM5 IHC={ihc_pct:.0f}% in 50–79% range — Gate 1 FAIL. "
                    f"CARMEN-LC03 IHC 50–79% PFS HR={CARMEN_IHC_50_79_PFS_HR} (HARMFUL)."
                )
            return GateStatus.FAIL, (
                f"CEACAM5 IHC={ihc_pct:.0f}% < 50% — Gate 1 FAIL. Below therapeutic threshold."
            )

        def _gate2_ccea(self, ccea_ngml):
            if ccea_ngml is None:
                return GateStatus.UNKNOWN, "cCEA not collected — Gate 2 unknown"
            if ccea_ngml >= CCEA_THRESHOLD:
                return GateStatus.PASS, (
                    f"cCEA={ccea_ngml:.0f} ng/mL ≥ {CCEA_THRESHOLD:.0f} ng/mL — Gate 2 PASS. "
                    f"cCEA ≥100 ORR: {CCEA_ORR_HIGH:.1%} vs {CCEA_ORR_LOW:.1%}, P=0.003."
                )
            return GateStatus.FAIL, (
                f"cCEA={ccea_ngml:.0f} ng/mL < {CCEA_THRESHOLD:.0f} ng/mL — Gate 2 FAIL."
            )

        def _overall(self, g1, g2):
            if g1 == GateStatus.PASS:
                return GateStatus.PASS, "Gate 1 (IHC ≥80%) PASS — primary gate met."
            if g2 == GateStatus.PASS:
                return GateStatus.PASS, (
                    "Gate 1 FAIL/UNKNOWN but Gate 2 (cCEA ≥100) PASS — surrogate gate met. "
                    "IHC ↔ cCEA Spearman ρ=0.43 (n=92) — moderate correlation."
                )
            if g1 == GateStatus.UNKNOWN and g2 == GateStatus.UNKNOWN:
                return GateStatus.UNKNOWN, "Both gates unknown — collect CEACAM5 IHC and cCEA."
            return GateStatus.FAIL, "Both gates FAIL — patient not eligible for CEACAM5-targeted therapy."

        def _panel(self, inp):
            panel = []
            if inp.ceacam5_ihc_pct is None:
                panel.append("CEACAM5 IHC (≥80% threshold) — Gate 1 primary")
            if inp.ccea_ngml is None:
                panel.append("cCEA serum (≥100 ng/mL threshold) — Gate 2 surrogate")
            if inp.ptmb is None:
                panel.append("pTMB GuardantOMNI (≥28 mut/Mb) — IO enrichment")
            if inp.immunoscore_ic is None:
                panel.append("Immunoscore-IC (HIGH/LOW) — TME inflammation")
            return panel if panel else ["All recommended biomarkers collected"]


    def evaluate_ceacam5_strategy(inp: Ceacam5Input) -> Ceacam5Result:
        """Convenience function."""
        return Ceacam5BiomarkerEngine().evaluate(inp)
    ''')

# ── TESTS: sl_therapy_bridge ─────────────────────────────────────────────────
write("tests/test_layer1/test_sl_therapy_bridge/__init__.py", "")
write("tests/test_layer1/test_sl_therapy_bridge/test_policy_engine.py", '''
    """Tests for SL Therapy Bridge — 20/20 policy tests."""
    import pytest
    from layer1_engines.sl_therapy_bridge import evaluate_sl_therapy_bridge, SLBridgeInput
    from layer1_engines.sl_therapy_bridge.models import PolicyDecision
    from layer1_engines.sl_therapy_bridge.policy_engine import PolicyEngine


    def _engine():
        return PolicyEngine()


    # ── ALLOW tests ────────────────────────────────────────────────────────────
    def test_brca2_olaparib_allow():
        rec = _engine().evaluate_pair("BRCA2_mut", "olaparib")
        assert rec.decision == PolicyDecision.ALLOW
        assert rec.evidence_tier == "FDA_APPROVED"


    def test_brca2_niraparib_allow():
        rec = _engine().evaluate_pair("BRCA2_mut", "niraparib")
        assert rec.decision == PolicyDecision.ALLOW


    def test_brca2_rucaparib_allow():
        rec = _engine().evaluate_pair("BRCA2_mut", "rucaparib")
        assert rec.decision == PolicyDecision.ALLOW


    def test_brca1_olaparib_allow():
        rec = _engine().evaluate_pair("BRCA1_mut", "olaparib")
        assert rec.decision == PolicyDecision.ALLOW


    def test_brca1_niraparib_allow():
        rec = _engine().evaluate_pair("BRCA1_mut", "niraparib")
        assert rec.decision == PolicyDecision.ALLOW


    def test_brca1_rucaparib_allow():
        rec = _engine().evaluate_pair("BRCA1_mut", "rucaparib")
        assert rec.decision == PolicyDecision.ALLOW


    def test_atm_ceralasertib_allow():
        rec = _engine().evaluate_pair("ATM_mut", "ceralasertib")
        assert rec.decision == PolicyDecision.ALLOW
        assert "40%" in rec.rationale  # RS-High ORR 40%


    def test_atm_berzosertib_allow():
        rec = _engine().evaluate_pair("ATM_mut", "berzosertib")
        assert rec.decision == PolicyDecision.ALLOW


    def test_arid1a_tazemetostat_allow():
        rec = _engine().evaluate_pair("ARID1A_mut", "tazemetostat")
        assert rec.decision == PolicyDecision.ALLOW


    def test_ccne1_cdk2_allow():
        rec = _engine().evaluate_pair("CCNE1_AMP", "CDK2_inhibitor")
        assert rec.decision == PolicyDecision.ALLOW


    def test_zeb1_itgav_allow():
        """BRIEF-1: ZEB1→ITGAV — CRISPR_VALIDATED, DEAL_READY."""
        rec = _engine().evaluate_pair("ZEB1", "ITGAV_inhibitor")
        assert rec.decision == PolicyDecision.ALLOW
        assert rec.evidence_tier == "CRISPR_VALIDATED"
        assert "−0.7184" in rec.rationale


    def test_spp1_nrf2_allow():
        """BRIEF-4: SPP1→NRF2 — CRISPR_VALIDATED."""
        rec = _engine().evaluate_pair("SPP1", "NRF2_inhibitor")
        assert rec.decision == PolicyDecision.ALLOW
        assert rec.evidence_tier == "CRISPR_VALIDATED"
        assert "−0.7326" in rec.rationale


    # ── ALLOW_PRECLINICAL_ONLY tests ───────────────────────────────────────────
    def test_smarca4_fhd909_preclinical_only():
        rec = _engine().evaluate_pair("SMARCA4_mut", "FHD-909")
        assert rec.decision == PolicyDecision.ALLOW_PRECLINICAL_ONLY
        assert rec.mandatory_disclosure is not None


    # ── WARN tests ─────────────────────────────────────────────────────────────
    def test_pten_null_mtor_warn():
        rec = _engine().evaluate_pair("PTEN_null", "mTOR_inhibitor")
        assert rec.decision == PolicyDecision.WARN
        assert rec.mandatory_disclosure is not None
        assert "1.82" in rec.mandatory_disclosure  # HARMFUL HR


    def test_pten_null_pi3k_warn():
        rec = _engine().evaluate_pair("PTEN_null", "PI3K_inhibitor")
        assert rec.decision == PolicyDecision.WARN


    # ── DISALLOW tests ─────────────────────────────────────────────────────────
    def test_kras_egfr_tki_disallow():
        rec = _engine().evaluate_pair("KRAS_mut", "EGFR_TKI")
        assert rec.decision == PolicyDecision.DISALLOW
        assert rec.evidence_tier == "FDA_CONTRAINDICATED"


    # ── BRIEF-2 EXCLUSION — CRITICAL ──────────────────────────────────────────
    def test_brief2_pten_null_itgav_not_in_map():
        """BRIEF-2 MUST return NOT_IN_MAP with mandatory disclosure."""
        rec = _engine().evaluate_pair("PTEN_null", "ITGAV")
        assert rec.decision == PolicyDecision.NOT_IN_MAP
        assert rec.mandatory_disclosure is not None
        assert "COMPUTATIONAL HYPOTHESIS" in rec.mandatory_disclosure
        assert "DepMap NULL" in rec.mandatory_disclosure


    def test_brief2_pten_null_itgav_inhibitor_not_in_map():
        """BRIEF-2 variant — also excluded."""
        rec = _engine().evaluate_pair("PTEN_null", "ITGAV_inhibitor")
        assert rec.decision == PolicyDecision.NOT_IN_MAP


    # ── Bridge integration test ────────────────────────────────────────────────
    def test_bridge_top_recommendation_prefers_crispr_validated():
        """ZEB1→ITGAV (CRISPR_VALIDATED) should rank above FDA_APPROVED in top rec."""
        inp = SLBridgeInput(
            patient_id="TEST-001",
            sl_partners=["ZEB1", "BRCA2_mut"],
            candidate_therapies=["ITGAV_inhibitor", "olaparib"],
        )
        result = evaluate_sl_therapy_bridge(inp)
        assert result.top_recommendation is not None
        assert result.top_recommendation.evidence_tier == "CRISPR_VALIDATED"
    ''')

# ── TESTS: ceacam5_biomarker_strategy ────────────────────────────────────────
write("tests/test_layer1/test_ceacam5_biomarker_strategy/__init__.py", "")
write("tests/test_layer1/test_ceacam5_biomarker_strategy/test_engine.py", '''
    """Tests for CEACAM5 Biomarker Strategy Engine."""
    import pytest
    from layer1_engines.ceacam5_biomarker_strategy import evaluate_ceacam5_strategy, Ceacam5Input
    from layer1_engines.ceacam5_biomarker_strategy.models import GateStatus
    from layer1_engines.ceacam5_biomarker_strategy.vector_scalpel import VECTOR_SCALPEL_RANKING


    def _inp(**kwargs):
        defaults = dict(patient_id="TEST-001", disease="MSS mCRC")
        defaults.update(kwargs)
        return Ceacam5Input(**defaults)


    def test_gate1_pass_ihc_ge80():
        result = evaluate_ceacam5_strategy(_inp(ceacam5_ihc_pct=85.0))
        assert result.gate_1_status == GateStatus.PASS
        assert result.overall_status == GateStatus.PASS


    def test_gate1_fail_ihc_50_79():
        """IHC 50–79% is HARMFUL per CARMEN-LC03 (HR=1.38)."""
        result = evaluate_ceacam5_strategy(_inp(ceacam5_ihc_pct=65.0))
        assert result.gate_1_status == GateStatus.FAIL
        assert "1.38" in result.gate_1_evidence  # HARMFUL HR


    def test_gate2_pass_ccea_ge100():
        result = evaluate_ceacam5_strategy(_inp(ccea_ngml=150.0))
        assert result.gate_2_status == GateStatus.PASS
        assert result.overall_status == GateStatus.PASS


    def test_gate2_fail_ccea_below_100():
        result = evaluate_ceacam5_strategy(_inp(ccea_ngml=50.0))
        assert result.gate_2_status == GateStatus.FAIL


    def test_both_gates_unknown():
        result = evaluate_ceacam5_strategy(_inp())
        assert result.gate_1_status == GateStatus.UNKNOWN
        assert result.gate_2_status == GateStatus.UNKNOWN
        assert result.overall_status == GateStatus.UNKNOWN


    def test_gate1_fail_gate2_pass_overall_pass():
        """Gate 2 surrogate can rescue Gate 1 failure."""
        result = evaluate_ceacam5_strategy(_inp(ceacam5_ihc_pct=30.0, ccea_ngml=120.0))
        assert result.gate_1_status == GateStatus.FAIL
        assert result.gate_2_status == GateStatus.PASS
        assert result.overall_status == GateStatus.PASS


    def test_open_assumption_oa08_always_present():
        result = evaluate_ceacam5_strategy(_inp())
        assert "OA-08" in result.open_assumption_oa08
        assert "28%" in result.open_assumption_oa08


    def test_vector_scalpel_ranking_order():
        """cCEA must rank #1, pTMB #2, IHC #3."""
        assert VECTOR_SCALPEL_RANKING[0].name == "cCEA"
        assert VECTOR_SCALPEL_RANKING[1].name == "pTMB"
        assert VECTOR_SCALPEL_RANKING[2].name == "IHC"


    def test_vector_scalpel_scores():
        """Verify exact scores from KG v2."""
        assert VECTOR_SCALPEL_RANKING[0].score == "27/33"
        assert VECTOR_SCALPEL_RANKING[1].score == "23/33"
        assert VECTOR_SCALPEL_RANKING[2].score == "20/33"


    def test_evidence_anchors_present():
        result = evaluate_ceacam5_strategy(_inp(ceacam5_ihc_pct=85.0))
        anchors = " ".join(result.evidence_anchors)
        assert "98.7%" in anchors   # Jansen 2024
        assert "79%" in anchors     # Warmerdam 2025
        assert "CARMEN" in anchors
        assert "PROCEADE" in anchors


    def test_ruo_flag():
        result = evaluate_ceacam5_strategy(_inp())
        assert result.ruo is True
    ''')

print("\nW2 complete — all files written.")
````
