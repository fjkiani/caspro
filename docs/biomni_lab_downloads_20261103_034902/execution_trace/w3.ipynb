````python

import os, textwrap

REPO = "/mnt/shared-workspace/shared/crispro-backend-v3-base"

def write(path, content):
    full = os.path.join(REPO, path)
    os.makedirs(os.path.dirname(full), exist_ok=True)
    with open(full, "w") as f:
        f.write(textwrap.dedent(content))
    print(f"  wrote {path} ({os.path.getsize(full)} bytes)")

def patch_file(path, old, new):
    full = os.path.join(REPO, path)
    with open(full, "r") as f:
        content = f.read()
    if old not in content:
        print(f"  WARN: patch target not found in {path}")
        return
    content = content.replace(old, new)
    with open(full, "w") as f:
        f.write(content)
    print(f"  patched {path}")

# ── LAYER 2: governance_gate.py ───────────────────────────────────────────────
write("layer2_gates/governance_gate.py", '''
    """
    Governance Gate — Cross-Cutting Claim Validator
    ================================================
    Enforces KG v2 governance guardrails across all pipeline outputs.
    Runs as a cross-cutting validator after all engines complete.

    Governance decisions (8D remediation, approved 2026-04-28):
      - PATH A formula: LOCKED (fit = clip((p·t) / ‖t‖₂, 0, 1))
      - PATH B: PROHIBITED in all outputs
      - LATIFY deltas: QUARANTINED (cite responder 0.8936 vs non-responder 0.6295 only)
      - GBM ZEB1 escape: PERMANENTLY CLOSED (3 datasets null/contradicted)
      - BRIEF-2 CRISPR validated: COMPUTATIONAL_HYPOTHESIS — mandatory disclosure
      - DDR 0.983: QUARANTINED — DL-07 publication-blocking
      - SC-001 BRAF WT: ACTIVE CONFLICT — CT26 carries BRAF V600E

    RUO — Research Use Only.
    """

    from dataclasses import dataclass, field
    from typing import Dict, List, Optional, Any


    # ── Blocked claims registry ────────────────────────────────────────────────
    BLOCKED_CLAIMS: Dict[str, str] = {
        "LATIFY_DELTA": (
            "QUARANTINED — cite responder 0.8936 vs non-responder 0.6295 only. "
            "Delta value is quarantined pending vector version resolution."
        ),
        "GBM_ZEB1_ESCAPE": (
            "PERMANENTLY CLOSED — 3 datasets null/contradicted: "
            "GLASS delta=−0.34 (p=0.000184), GSE222515 delta=−0.431 (p=0.0334), "
            "GSE174554 delta=−0.076 (p=0.742)."
        ),
        "BRIEF_2_CRISPR_VALIDATED": (
            "COMPUTATIONAL_HYPOTHESIS — DepMap NULL across all 3 proxies. "
            "Mandatory disclosure required in every context."
        ),
        "DDR_0983": (
            "QUARANTINED — DL-07 publication-blocking. "
            "Do not cite DDR 0.983 figure until resolved."
        ),
        "PATH_B_FORMULA": (
            "PROHIBITED — use PATH A only. "
            "PATH A: fit = clip((p·t) / ‖t‖₂, 0, 1). "
            "PATH B (cosine similarity with ‖p‖₂ in denominator) is prohibited."
        ),
        "SC_001_BRAF_WT": (
            "ACTIVE CONFLICT — CT26 carries BRAF V600E (ATCC/CCLE/COSMIC). "
            "BreAK CRC-001 Exclusion Criterion #7 requires BRAF wild-type patients. "
            "Must be resolved before enrollment."
        ),
    }

    # ── Mandatory disclosures ──────────────────────────────────────────────────
    MANDATORY_DISCLOSURES: Dict[str, str] = {
        "BRIEF_2": (
            "PTEN-null → ITGAV is a COMPUTATIONAL HYPOTHESIS. "
            "DepMap NULL across all 3 proxies. Not validated by CRISPR."
        ),
        "OV_ZEB1_ARM_INTERACTION": (
            "ZEB1 is PROGNOSTIC, NOT ARM-SPECIFIC. "
            "Arm interaction p=0.246 — not statistically significant."
        ),
        "GATE_1_FRACTION": (
            "28% CEACAM5 IHC ≥80% fraction is OPEN_ASSUMPTION OA-08. "
            "Not yet validated in a prospective MSS mCRC cohort."
        ),
        "LATIFY_RESPONDER_DELTA": (
            "LATIFY delta is QUARANTINED. "
            "Cite responder 0.8936 vs non-responder 0.6295 only."
        ),
    }

    # ── Permanently closed programs ────────────────────────────────────────────
    PERMANENTLY_CLOSED: Dict[str, str] = {
        "GBM": (
            "GBM ZEB1 escape program PERMANENTLY CLOSED. "
            "3 datasets null/contradicted: GLASS, GSE222515, GSE174554."
        ),
    }

    # ── Quarantined items ──────────────────────────────────────────────────────
    QUARANTINED: Dict[str, str] = {
        "LATIFY_DELTA": "Vector version unresolved — CT-03/DL-05/PC-06 quarantined.",
        "DDR_0983": "DL-07 publication-blocking — do not cite.",
        "BRIEF_2_CRISPR": "DepMap NULL — COMPUTATIONAL_HYPOTHESIS only.",
    }


    @dataclass
    class GovernanceViolation:
        claim_key: str
        severity: str           # "BLOCKED" | "QUARANTINED" | "MANDATORY_DISCLOSURE"
        message: str
        remediation: str


    @dataclass
    class GovernanceGateResult:
        passed: bool
        violations: List[GovernanceViolation]
        mandatory_disclosures: List[str]
        path_a_confirmed: bool
        ruo: bool = True

        @property
        def violation_count(self) -> int:
            return len(self.violations)

        @property
        def has_blocked_claims(self) -> bool:
            return any(v.severity == "BLOCKED" for v in self.violations)


    class GovernanceGate:
        """
        Cross-cutting governance validator.
        Checks pipeline outputs for blocked claims, quarantined items,
        and mandatory disclosures.
        """

        def evaluate(
            self,
            output_text: str = "",
            claim_keys: Optional[List[str]] = None,
            check_path_a: bool = True,
        ) -> GovernanceGateResult:
            """
            Evaluate governance compliance.

            Args:
                output_text: Free-text output to scan for blocked claim patterns.
                claim_keys: Explicit claim keys to check (e.g. ["LATIFY_DELTA"]).
                check_path_a: Whether to verify PATH A formula is referenced.

            Returns:
                GovernanceGateResult with violations and mandatory disclosures.
            """
            violations = []
            disclosures = []

            # Check explicit claim keys
            if claim_keys:
                for key in claim_keys:
                    if key in BLOCKED_CLAIMS:
                        violations.append(GovernanceViolation(
                            claim_key=key,
                            severity="BLOCKED",
                            message=BLOCKED_CLAIMS[key],
                            remediation=f"Remove or replace claim '{key}' per governance guardrails.",
                        ))
                    if key in QUARANTINED:
                        violations.append(GovernanceViolation(
                            claim_key=key,
                            severity="QUARANTINED",
                            message=QUARANTINED[key],
                            remediation=f"Quarantine claim '{key}' — do not publish.",
                        ))
                    if key in MANDATORY_DISCLOSURES:
                        disclosures.append(MANDATORY_DISCLOSURES[key])

            # Scan output text for prohibited patterns
            text_lower = output_text.lower()
            if "path b" in text_lower or "cosine similarity" in text_lower:
                violations.append(GovernanceViolation(
                    claim_key="PATH_B_FORMULA",
                    severity="BLOCKED",
                    message=BLOCKED_CLAIMS["PATH_B_FORMULA"],
                    remediation="Replace PATH B formula with PATH A: fit = clip((p·t) / ‖t‖₂, 0, 1).",
                ))
            if "gbm" in text_lower and "zeb1" in text_lower and "escape" in text_lower:
                violations.append(GovernanceViolation(
                    claim_key="GBM_ZEB1_ESCAPE",
                    severity="BLOCKED",
                    message=BLOCKED_CLAIMS["GBM_ZEB1_ESCAPE"],
                    remediation="Remove GBM ZEB1 escape claim — program permanently closed.",
                ))
            if "brief-2" in text_lower or "brief_2" in text_lower:
                if "computational hypothesis" not in text_lower:
                    disclosures.append(MANDATORY_DISCLOSURES["BRIEF_2"])
            if "ddr" in text_lower and "0.983" in output_text:
                violations.append(GovernanceViolation(
                    claim_key="DDR_0983",
                    severity="QUARANTINED",
                    message=BLOCKED_CLAIMS["DDR_0983"],
                    remediation="Remove DDR 0.983 figure — quarantined pending DL-07 resolution.",
                ))

            # PATH A confirmation
            path_a_confirmed = (
                "path a" in text_lower or
                "path_a" in text_lower or
                not check_path_a  # skip if not requested
            )

            passed = not any(v.severity == "BLOCKED" for v in violations)

            return GovernanceGateResult(
                passed=passed,
                violations=violations,
                mandatory_disclosures=disclosures,
                path_a_confirmed=path_a_confirmed,
            )


    # Module-level convenience
    _gate = GovernanceGate()


    def evaluate_governance(
        output_text: str = "",
        claim_keys: Optional[List[str]] = None,
        check_path_a: bool = True,
    ) -> GovernanceGateResult:
        """Convenience function."""
        return _gate.evaluate(output_text, claim_keys, check_path_a)
    ''')

# ── LAYER 1: matching_rules_engine/__init__.py ────────────────────────────────
write("layer1_engines/matching_rules_engine/__init__.py", '''
    """Matching Rules Engine — 25-rule patient-trial matching."""
    from .engine import MatchingRulesEngine, evaluate_matching_rules
    from .models import MatchingInput, MatchingResult, RuleResult, ConfidenceTier
    from .rules import MATCHING_RULES

    __all__ = [
        "MatchingRulesEngine", "evaluate_matching_rules",
        "MatchingInput", "MatchingResult", "RuleResult", "ConfidenceTier",
        "MATCHING_RULES",
    ]
    ''')

# ── LAYER 1: matching_rules_engine/models.py ─────────────────────────────────
write("layer1_engines/matching_rules_engine/models.py", '''
    """Matching Rules Engine — Data Models. RUO."""

    from dataclasses import dataclass, field
    from typing import List, Optional
    from enum import Enum


    class ConfidenceTier(str, Enum):
        HIGH = "HIGH"
        MODERATE = "MODERATE"
        CONDITIONAL = "CONDITIONAL"


    @dataclass
    class RuleResult:
        rule_id: str
        rule_name: str
        fired: bool
        confidence: ConfidenceTier
        rationale: str
        evidence_anchor: str


    @dataclass
    class MatchingInput:
        patient_id: str
        disease: str
        mss_confirmed: bool = False
        liver_metastasis: Optional[bool] = None
        ptmb: Optional[float] = None
        ceacam5_ihc_pct: Optional[float] = None
        ccea_ngml: Optional[float] = None
        immunoscore_ic: Optional[str] = None
        brca_status: Optional[str] = None      # "BRCA1_mut" / "BRCA2_mut" / "WT"
        atm_status: Optional[str] = None       # "ATM_mut" / "WT"
        kras_status: Optional[str] = None      # "KRAS_mut" / "WT"
        pten_status: Optional[str] = None      # "PTEN_null" / "intact"
        zeb1_expression: Optional[str] = None  # "HIGH" / "LOW"
        fit_score: Optional[float] = None      # PATH A fit score


    @dataclass
    class MatchingResult:
        patient_id: str
        rules_fired: List[RuleResult]
        rules_not_fired: List[RuleResult]
        high_confidence_count: int
        moderate_confidence_count: int
        conditional_count: int
        top_trial_matches: List[str]
        overall_match_score: float
        ruo: bool = True
    ''')

# ── LAYER 1: matching_rules_engine/rules.py ───────────────────────────────────
write("layer1_engines/matching_rules_engine/rules.py", '''
    """
    Matching Rules — 25 rules from KG v2 matching_rules_v2.

    Rules are organized by confidence tier:
      HIGH: Strong evidence from randomized data
      MODERATE: Evidence from non-randomized or exploratory data
      CONDITIONAL: Mechanistic rationale, limited clinical evidence

    RUO — Research Use Only.
    """

    from dataclasses import dataclass
    from typing import Callable, Optional
    from .models import MatchingInput, ConfidenceTier


    @dataclass
    class Rule:
        rule_id: str
        name: str
        confidence: ConfidenceTier
        evidence_anchor: str
        condition: Callable[[MatchingInput], bool]
        rationale_fn: Callable[[MatchingInput], str]


    def _r(rule_id, name, confidence, evidence_anchor, condition, rationale_fn):
        return Rule(rule_id, name, confidence, evidence_anchor, condition, rationale_fn)


    MATCHING_RULES = [
        # ── HIGH confidence rules ──────────────────────────────────────────────
        _r("R-01", "MSS CRC + NLM → IO permissive",
           ConfidenceTier.HIGH,
           "CO.26: no-LM PFS HR=0.54 (Pint=0.02); REGONIVO no-LM ORR=22%",
           lambda p: p.mss_confirmed and p.liver_metastasis is False,
           lambda p: "MSS CRC + no liver metastases → IO permissive. CO.26 no-LM PFS HR=0.54."),

        _r("R-02", "MSS CRC + liver mets → IO suppressed",
           ConfidenceTier.HIGH,
           "CO.26: liver-met PFS HR=1.39 (HARMFUL); REGONIVO 0/5 liver-met responders",
           lambda p: p.mss_confirmed and p.liver_metastasis is True,
           lambda p: "MSS CRC + liver metastases → IO suppressed. CO.26 liver-met PFS HR=1.39 (HARMFUL)."),

        _r("R-03", "pTMB ≥28 → IO enriched",
           ConfidenceTier.HIGH,
           "CO.26: pTMB≥28 OS HR=0.34 (90% CI 0.18–0.63, p=0.022)",
           lambda p: p.ptmb is not None and p.ptmb >= 28.0,
           lambda p: f"pTMB={p.ptmb:.1f} ≥28 mut/Mb → IO enriched. CO.26 OS HR=0.34."),

        _r("R-04", "pTMB <28 → IO not enriched",
           ConfidenceTier.HIGH,
           "CO.26: pTMB≥28 OS HR=0.34; below-threshold no IO benefit",
           lambda p: p.ptmb is not None and p.ptmb < 28.0,
           lambda p: f"pTMB={p.ptmb:.1f} <28 mut/Mb → IO not enriched."),

        _r("R-05", "BRCA1/2 mut → PARPi eligible",
           ConfidenceTier.HIGH,
           "FDA-approved: olaparib, niraparib, rucaparib for BRCA1/2-mutated cancers",
           lambda p: p.brca_status in ("BRCA1_mut", "BRCA2_mut"),
           lambda p: f"{p.brca_status} → PARPi eligible (olaparib/niraparib/rucaparib). FDA-approved."),

        _r("R-06", "KRAS mut → EGFR TKI contraindicated",
           ConfidenceTier.HIGH,
           "FDA label: KRAS mutation = primary resistance to EGFR TKIs in CRC",
           lambda p: p.kras_status == "KRAS_mut",
           lambda p: "KRAS_mut → EGFR TKI contraindicated. FDA label."),

        _r("R-07", "CEACAM5 IHC ≥80% → CEACAM5-targeted therapy eligible",
           ConfidenceTier.HIGH,
           "CARMEN-LC03: IHC ≥80% OS HR=0.71 (post-hoc); PROCEADE-CRC-01: ORR 7.5%",
           lambda p: p.ceacam5_ihc_pct is not None and p.ceacam5_ihc_pct >= 80.0,
           lambda p: f"CEACAM5 IHC={p.ceacam5_ihc_pct:.0f}% ≥80% → CEACAM5-targeted therapy eligible."),

        _r("R-08", "CEACAM5 IHC 50–79% → CEACAM5-targeted therapy HARMFUL",
           ConfidenceTier.HIGH,
           "CARMEN-LC03: IHC 50–79% PFS HR=1.38 (HARMFUL)",
           lambda p: p.ceacam5_ihc_pct is not None and 50.0 <= p.ceacam5_ihc_pct < 80.0,
           lambda p: f"CEACAM5 IHC={p.ceacam5_ihc_pct:.0f}% in 50–79% range → HARMFUL. CARMEN-LC03 HR=1.38."),

        _r("R-09", "cCEA ≥100 → CEACAM5 surrogate gate PASS",
           ConfidenceTier.HIGH,
           "cCEA ≥100 ORR in CEACAM5 HE NSCLC: 41.7% vs 8.1%, P=0.003",
           lambda p: p.ccea_ngml is not None and p.ccea_ngml >= 100.0,
           lambda p: f"cCEA={p.ccea_ngml:.0f} ng/mL ≥100 → CEACAM5 surrogate gate PASS."),

        _r("R-10", "ATM mut → ATR inhibitor eligible",
           ConfidenceTier.HIGH,
           "RS-High ORR 40% vs RS-Low 5% (KG v2 ATR/DDR corpus)",
           lambda p: p.atm_status == "ATM_mut",
           lambda p: "ATM_mut → ATR inhibitor eligible (ceralasertib/berzosertib). RS-High ORR 40%."),

        # ── MODERATE confidence rules ──────────────────────────────────────────
        _r("R-11", "ZEB1 high → ITGAV inhibitor eligible (BRIEF-1)",
           ConfidenceTier.MODERATE,
           "BRIEF-1: delta=−0.7184, FDR=0.001203, CRISPR_VALIDATED, DEAL_READY",
           lambda p: p.zeb1_expression == "HIGH",
           lambda p: "ZEB1 HIGH → ITGAV inhibitor eligible. BRIEF-1: delta=−0.7184, CRISPR_VALIDATED."),

        _r("R-12", "Immunoscore-IC HIGH → IO benefit likely",
           ConfidenceTier.MODERATE,
           "AtezoTRIBE: Immunoscore-IC high HR ~0.50 (exploratory)",
           lambda p: p.immunoscore_ic == "HIGH",
           lambda p: "Immunoscore-IC HIGH → IO benefit likely. AtezoTRIBE HR ~0.50 (exploratory)."),

        _r("R-13", "PATH A fit ≥0.70 → trial match MODERATE-HIGH",
           ConfidenceTier.MODERATE,
           "PATH A formula: fit = clip((p·t) / ‖t‖₂, 0, 1) (LOCKED 2026-04-28)",
           lambda p: p.fit_score is not None and p.fit_score >= 0.70,
           lambda p: f"PATH A fit={p.fit_score:.4f} ≥0.70 → trial match MODERATE-HIGH."),

        _r("R-14", "PATH A fit <0.40 → trial match LOW",
           ConfidenceTier.MODERATE,
           "PATH A formula: fit = clip((p·t) / ‖t‖₂, 0, 1) (LOCKED 2026-04-28)",
           lambda p: p.fit_score is not None and p.fit_score < 0.40,
           lambda p: f"PATH A fit={p.fit_score:.4f} <0.40 → trial match LOW."),

        _r("R-15", "MSS CRC + NLM + pTMB≥28 → all three IO conditions met",
           ConfidenceTier.MODERATE,
           "CO.26 combined analysis: NLM + pTMB≥28 = strongest IO enrichment",
           lambda p: (p.mss_confirmed and p.liver_metastasis is False
                      and p.ptmb is not None and p.ptmb >= 28.0),
           lambda p: "MSS CRC + NLM + pTMB≥28 → all three IO conditions met. Optimal IO candidate."),

        _r("R-16", "PTEN null + mTOR inhibitor → WARN (mixed evidence)",
           ConfidenceTier.MODERATE,
           "Adavosertib: PTEN-loss PFS HR=1.82 (HARMFUL); mTOR inhibitor benefit not established",
           lambda p: p.pten_status == "PTEN_null",
           lambda p: "PTEN_null → mTOR/PI3K inhibitor WARN. Adavosertib PTEN-loss HR=1.82 (HARMFUL)."),

        _r("R-17", "CEACAM5 any positivity in CRC → 98.7% prevalence",
           ConfidenceTier.MODERATE,
           "Jansen et al. Cancers 2024: 98.7% any CEACAM5 positivity in CRC",
           lambda p: "crc" in p.disease.lower() or "colorectal" in p.disease.lower(),
           lambda p: "CRC disease → CEACAM5 any positivity 98.7% (Jansen 2024). Test IHC ≥80%."),

        _r("R-18", "CRLM → CEACAM5 positivity 79%",
           ConfidenceTier.MODERATE,
           "Warmerdam et al. EJNMMI Res 2025: 79% CEACAM5 positivity in CRLM",
           lambda p: ("crlm" in p.disease.lower() or
                      ("crc" in p.disease.lower() and p.liver_metastasis is True)),
           lambda p: "CRLM → CEACAM5 positivity 79% (Warmerdam 2025). Test IHC ≥80%."),

        _r("R-19", "MSS CRC + no biomarker gate → QUILT-2.004 failure risk",
           ConfidenceTier.MODERATE,
           "QUILT-2.004: PFS HR=1.061 (p=0.91) — TERMINATED in unselected MSS CRC",
           lambda p: p.mss_confirmed and p.ptmb is None and p.immunoscore_ic is None,
           lambda p: "MSS CRC + no biomarker gate → QUILT-2.004 failure risk (HR=1.061, p=0.91)."),

        _r("R-20", "BreAK CRC-001 fit score 0.7375 → MODERATE-HIGH match",
           ConfidenceTier.MODERATE,
           "PATH A derivation: dot=0.685, norm=0.9288, fit=0.7375 (KG v2 locked)",
           lambda p: (p.fit_score is not None and
                      abs(p.fit_score - 0.7375) < 0.01 and
                      p.mss_confirmed),
           lambda p: "BreAK CRC-001 canonical fit=0.7375 (MODERATE-HIGH). io=46.7%, mapk=30.7%."),

        # ── CONDITIONAL confidence rules ───────────────────────────────────────
        _r("R-21", "ZEB1 high + OV → prognostic (not arm-specific)",
           ConfidenceTier.CONDITIONAL,
           "EscapeMap OV: ZEB1 arm interaction p=0.246 — PROGNOSTIC_NOT_ARM_SPECIFIC",
           lambda p: p.zeb1_expression == "HIGH" and "ov" in p.disease.lower(),
           lambda p: "ZEB1 HIGH in OV → PROGNOSTIC only (arm interaction p=0.246). Not arm-specific."),

        _r("R-22", "CEACAM5 IHC ≥80% + NLM → optimal CEACAM5 candidate",
           ConfidenceTier.CONDITIONAL,
           "CEACAM5 IHC ≥80% + NLM = two-gate PASS + IO permissive",
           lambda p: (p.ceacam5_ihc_pct is not None and p.ceacam5_ihc_pct >= 80.0
                      and p.liver_metastasis is False),
           lambda p: "CEACAM5 IHC ≥80% + NLM → optimal CEACAM5 candidate. Both gates met."),

        _r("R-23", "pTMB ≥28 + NLM + Immunoscore HIGH → three-conditions IO optimal",
           ConfidenceTier.CONDITIONAL,
           "CO.26 + AtezoTRIBE combined: three-conditions framework",
           lambda p: (p.ptmb is not None and p.ptmb >= 28.0
                      and p.liver_metastasis is False
                      and p.immunoscore_ic == "HIGH"),
           lambda p: "pTMB≥28 + NLM + Immunoscore HIGH → three-conditions IO optimal."),

        _r("R-24", "BRCA1/2 mut + ATM mut → DDR-high profile",
           ConfidenceTier.CONDITIONAL,
           "DDR-high profile: PARPi + ATR inhibitor combination potential",
           lambda p: (p.brca_status in ("BRCA1_mut", "BRCA2_mut")
                      and p.atm_status == "ATM_mut"),
           lambda p: "BRCA1/2 mut + ATM mut → DDR-high profile. PARPi + ATR inhibitor combination."),

        _r("R-25", "ZEB1 high + pTMB ≥28 → BRIEF-1 + IO combination candidate",
           ConfidenceTier.CONDITIONAL,
           "BRIEF-1 (ZEB1→ITGAV) + CO.26 pTMB≥28 IO enrichment",
           lambda p: (p.zeb1_expression == "HIGH"
                      and p.ptmb is not None and p.ptmb >= 28.0),
           lambda p: "ZEB1 HIGH + pTMB≥28 → BRIEF-1 + IO combination candidate."),
    ]

    assert len(MATCHING_RULES) == 25, f"Expected 25 rules, got {len(MATCHING_RULES)}"
    ''')

# ── LAYER 1: matching_rules_engine/engine.py ─────────────────────────────────
write("layer1_engines/matching_rules_engine/engine.py", '''
    """Matching Rules Engine — evaluates all 25 rules. RUO."""

    from typing import List
    from .models import MatchingInput, MatchingResult, RuleResult, ConfidenceTier
    from .rules import MATCHING_RULES


    class MatchingRulesEngine:
        """Evaluates all 25 matching rules against a patient profile."""

        def evaluate(self, inp: MatchingInput) -> MatchingResult:
            fired, not_fired = [], []
            for rule in MATCHING_RULES:
                try:
                    fires = rule.condition(inp)
                except Exception:
                    fires = False
                rationale = rule.rationale_fn(inp) if fires else f"Rule {rule.rule_id} condition not met."
                rr = RuleResult(
                    rule_id=rule.rule_id,
                    rule_name=rule.name,
                    fired=fires,
                    confidence=rule.confidence,
                    rationale=rationale,
                    evidence_anchor=rule.evidence_anchor,
                )
                (fired if fires else not_fired).append(rr)

            high = sum(1 for r in fired if r.confidence == ConfidenceTier.HIGH)
            moderate = sum(1 for r in fired if r.confidence == ConfidenceTier.MODERATE)
            conditional = sum(1 for r in fired if r.confidence == ConfidenceTier.CONDITIONAL)

            score = min(1.0, (high * 0.5 + moderate * 0.3 + conditional * 0.2) / max(len(MATCHING_RULES) * 0.5, 1))
            top_trials = self._top_trials(fired)

            return MatchingResult(
                patient_id=inp.patient_id,
                rules_fired=fired,
                rules_not_fired=not_fired,
                high_confidence_count=high,
                moderate_confidence_count=moderate,
                conditional_count=conditional,
                top_trial_matches=top_trials,
                overall_match_score=round(score, 4),
            )

        def _top_trials(self, fired: List[RuleResult]) -> List[str]:
            trials = []
            fired_ids = {r.rule_id for r in fired}
            if "R-01" in fired_ids and "R-03" in fired_ids:
                trials.append("BreAK CRC-001 (STC-1010) — NLM + pTMB≥28 optimal")
            if "R-05" in fired_ids:
                trials.append("PARPi trial (olaparib/niraparib/rucaparib) — BRCA1/2 mut")
            if "R-07" in fired_ids:
                trials.append("CEACAM5-targeted trial — IHC ≥80% gate PASS")
            if "R-10" in fired_ids:
                trials.append("ATR inhibitor trial (ceralasertib/berzosertib) — ATM mut")
            if "R-11" in fired_ids:
                trials.append("ITGAV inhibitor trial — ZEB1 HIGH (BRIEF-1)")
            return trials if trials else ["No high-confidence trial match identified"]


    def evaluate_matching_rules(inp: MatchingInput) -> MatchingResult:
        """Convenience function."""
        return MatchingRulesEngine().evaluate(inp)
    ''')

# ── FIX: clinical_trials_routes.py — 8D STC-1010 vector ─────────────────────
ct_path = os.path.join(REPO, "layer4_output/clinical_trials_routes.py")
if os.path.exists(ct_path):
    with open(ct_path, "r") as f:
        ct_content = f.read()
    # Fix 6D → 8D vector
    old_vec = "_STC1010_VECTOR = [0.1, 0.15, 0.05, 0.1, 0.9, 0.0]"
    new_vec = (
        "# STC-1010 8D vector (CORRECTED from 6D — KG v2 locked 2026-04-28)\n"
        "# Axes: ddr=0.10, mapk=0.30, pi3k=0.10, vegf=0.20, her2=0.00, io=0.80, efflux=0.10, rss=0.10\n"
        "# io=0.80 is primary axis (46.7% of dot product); mapk=0.30 secondary (30.7%)\n"
        "_STC1010_VECTOR = [0.10, 0.30, 0.10, 0.20, 0.00, 0.80, 0.10, 0.10]  # 8D: ddr,mapk,pi3k,vegf,her2,io,efflux,rss"
    )
    if old_vec in ct_content:
        ct_content = ct_content.replace(old_vec, new_vec)
        with open(ct_path, "w") as f:
            f.write(ct_content)
        print("  patched clinical_trials_routes.py — 8D STC-1010 vector")
    else:
        # Try to find any 6D vector pattern
        import re
        pattern = r'_STC1010_VECTOR\s*=\s*\[.*?\]'
        match = re.search(pattern, ct_content)
        if match:
            ct_content = ct_content[:match.start()] + new_vec + ct_content[match.end():]
            with open(ct_path, "w") as f:
                f.write(ct_content)
            print("  patched clinical_trials_routes.py — 8D STC-1010 vector (regex)")
        else:
            print(f"  WARN: STC1010_VECTOR not found in clinical_trials_routes.py — appending fix")
            with open(ct_path, "a") as f:
                f.write(f"\n\n# KG v2 CORRECTION (2026-04-28)\n{new_vec}\n")
else:
    print(f"  WARN: clinical_trials_routes.py not found at {ct_path}")

# ── FIX: mechanism_fit_ranker.py — PATH A label ───────────────────────────────
mfr_path = os.path.join(REPO, "layer1_engines/mechanism_fit/mechanism_fit_ranker.py")
if os.path.exists(mfr_path):
    with open(mfr_path, "r") as f:
        mfr_content = f.read()
    # Add PATH A provenance constant after the first import block
    path_a_const = (
        '\n# PATH A formula provenance (LOCKED 2026-04-28)\n'
        'PATH_A_FORMULA_VERSION = "PATH_A_LOCKED_2026-04-28"\n'
        'PATH_A_FORMULA = "fit = clip((p·t) / ‖t‖₂, 0, 1)"\n'
        'PATH_A_APPROVAL = "Approved by Fahad Kiani 2026-04-28. PATH B prohibited."\n'
        'VECTOR_VERSION_8D = "8D.v1"  # axes: ddr,mapk,pi3k,vegf,her2,io,efflux,rss\n'
    )
    if "PATH_A_FORMULA_VERSION" not in mfr_content:
        # Insert after the last import line
        lines = mfr_content.split('\n')
        last_import = 0
        for i, line in enumerate(lines):
            if line.startswith('import ') or line.startswith('from '):
                last_import = i
        lines.insert(last_import + 1, path_a_const)
        with open(mfr_path, "w") as f:
            f.write('\n'.join(lines))
        print("  patched mechanism_fit_ranker.py — PATH A label added")
    else:
        print("  mechanism_fit_ranker.py already has PATH_A_FORMULA_VERSION")
else:
    print(f"  WARN: mechanism_fit_ranker.py not found at {mfr_path}")

# ── TESTS: governance_gate ────────────────────────────────────────────────────
write("tests/test_layer2/test_governance_gate.py", '''
    """Tests for Governance Gate."""
    import pytest
    from layer2_gates.governance_gate import (
        evaluate_governance, GovernanceGate,
        BLOCKED_CLAIMS, MANDATORY_DISCLOSURES, QUARANTINED,
    )


    def test_path_b_blocked():
        result = evaluate_governance(output_text="Using PATH B cosine similarity formula")
        assert not result.passed
        assert any(v.claim_key == "PATH_B_FORMULA" for v in result.violations)


    def test_gbm_zeb1_escape_blocked():
        result = evaluate_governance(output_text="GBM ZEB1 escape program shows promise")
        assert not result.passed
        assert any(v.claim_key == "GBM_ZEB1_ESCAPE" for v in result.violations)


    def test_ddr_0983_quarantined():
        result = evaluate_governance(output_text="DDR score is 0.983 in this cohort")
        assert any(v.claim_key == "DDR_0983" for v in result.violations)


    def test_brief2_without_disclosure_triggers_disclosure():
        result = evaluate_governance(output_text="BRIEF-2 shows strong SL signal")
        assert any("COMPUTATIONAL HYPOTHESIS" in d for d in result.mandatory_disclosures)


    def test_brief2_with_disclosure_no_violation():
        result = evaluate_governance(
            output_text="BRIEF-2 is a computational hypothesis — DepMap NULL"
        )
        # Should not add disclosure since it's already present
        assert not any(v.claim_key == "BRIEF_2_CRISPR_VALIDATED" for v in result.violations)


    def test_explicit_claim_key_latify_delta():
        result = evaluate_governance(claim_keys=["LATIFY_DELTA"])
        assert any(v.claim_key == "LATIFY_DELTA" for v in result.violations)


    def test_explicit_claim_key_mandatory_disclosure():
        result = evaluate_governance(claim_keys=["BRIEF_2"])
        assert any("COMPUTATIONAL HYPOTHESIS" in d for d in result.mandatory_disclosures)


    def test_clean_output_passes():
        result = evaluate_governance(
            output_text="PATH A fit score is 0.7375 for BreAK CRC-001.",
            check_path_a=True,
        )
        assert result.passed
        assert result.path_a_confirmed


    def test_25_blocked_claims_and_disclosures_defined():
        """Verify governance registry completeness."""
        assert len(BLOCKED_CLAIMS) >= 6
        assert len(MANDATORY_DISCLOSURES) >= 4
        assert len(QUARANTINED) >= 3


    def test_ruo_flag():
        result = evaluate_governance()
        assert result.ruo is True
    ''')

# ── TESTS: matching_rules_engine ─────────────────────────────────────────────
write("tests/test_layer1/test_matching_rules_engine/__init__.py", "")
write("tests/test_layer1/test_matching_rules_engine/test_engine.py", '''
    """Tests for Matching Rules Engine — 25 rules."""
    import pytest
    from layer1_engines.matching_rules_engine import evaluate_matching_rules, MatchingInput
    from layer1_engines.matching_rules_engine.models import ConfidenceTier
    from layer1_engines.matching_rules_engine.rules import MATCHING_RULES


    def test_exactly_25_rules():
        assert len(MATCHING_RULES) == 25


    def _inp(**kwargs):
        defaults = dict(patient_id="TEST-001", disease="MSS mCRC")
        defaults.update(kwargs)
        return MatchingInput(**defaults)


    def test_nlm_fires_r01():
        result = evaluate_matching_rules(_inp(mss_confirmed=True, liver_metastasis=False))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-01" in fired_ids


    def test_liver_met_fires_r02():
        result = evaluate_matching_rules(_inp(mss_confirmed=True, liver_metastasis=True))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-02" in fired_ids


    def test_ptmb_ge28_fires_r03():
        result = evaluate_matching_rules(_inp(ptmb=35.0))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-03" in fired_ids


    def test_ptmb_lt28_fires_r04():
        result = evaluate_matching_rules(_inp(ptmb=15.0))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-04" in fired_ids


    def test_brca2_fires_r05():
        result = evaluate_matching_rules(_inp(brca_status="BRCA2_mut"))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-05" in fired_ids


    def test_kras_fires_r06():
        result = evaluate_matching_rules(_inp(kras_status="KRAS_mut"))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-06" in fired_ids


    def test_ceacam5_ge80_fires_r07():
        result = evaluate_matching_rules(_inp(ceacam5_ihc_pct=85.0))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-07" in fired_ids


    def test_ceacam5_50_79_fires_r08():
        result = evaluate_matching_rules(_inp(ceacam5_ihc_pct=65.0))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-08" in fired_ids


    def test_atm_fires_r10():
        result = evaluate_matching_rules(_inp(atm_status="ATM_mut"))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-10" in fired_ids


    def test_zeb1_high_fires_r11():
        result = evaluate_matching_rules(_inp(zeb1_expression="HIGH"))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-11" in fired_ids


    def test_fit_score_ge70_fires_r13():
        result = evaluate_matching_rules(_inp(fit_score=0.7375))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-13" in fired_ids


    def test_break_crc_canonical_fires_r20():
        result = evaluate_matching_rules(_inp(
            mss_confirmed=True, fit_score=0.7375
        ))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-20" in fired_ids


    def test_three_conditions_fires_r15():
        result = evaluate_matching_rules(_inp(
            mss_confirmed=True, liver_metastasis=False, ptmb=35.0
        ))
        fired_ids = {r.rule_id for r in result.rules_fired}
        assert "R-15" in fired_ids


    def test_top_trial_matches_break_crc():
        result = evaluate_matching_rules(_inp(
            mss_confirmed=True, liver_metastasis=False, ptmb=35.0
        ))
        assert any("BreAK" in t for t in result.top_trial_matches)


    def test_ruo_flag():
        result = evaluate_matching_rules(_inp())
        assert result.ruo is True
    ''')

# ── TESTS: 8D vector fix ──────────────────────────────────────────────────────
write("tests/test_layer4/test_stc1010_vector_fix.py", '''
    """Verify STC-1010 vector is 8D and correct."""
    import pytest


    def test_stc1010_vector_is_8d():
        """STC-1010 vector must be 8D [ddr,mapk,pi3k,vegf,her2,io,efflux,rss]."""
        try:
            from layer4_output.clinical_trials_routes import _STC1010_VECTOR
        except ImportError:
            pytest.skip("clinical_trials_routes not importable in test environment")

        assert len(_STC1010_VECTOR) == 8, (
            f"STC-1010 vector must be 8D, got {len(_STC1010_VECTOR)}D: {_STC1010_VECTOR}"
        )


    def test_stc1010_vector_io_axis():
        """io axis (index 5) must be 0.80 — primary axis."""
        try:
            from layer4_output.clinical_trials_routes import _STC1010_VECTOR
        except ImportError:
            pytest.skip("clinical_trials_routes not importable in test environment")

        assert abs(_STC1010_VECTOR[5] - 0.80) < 0.001, (
            f"io axis (index 5) must be 0.80, got {_STC1010_VECTOR[5]}"
        )


    def test_stc1010_vector_mapk_axis():
        """mapk axis (index 1) must be 0.30 — secondary axis."""
        try:
            from layer4_output.clinical_trials_routes import _STC1010_VECTOR
        except ImportError:
            pytest.skip("clinical_trials_routes not importable in test environment")

        assert abs(_STC1010_VECTOR[1] - 0.30) < 0.001, (
            f"mapk axis (index 1) must be 0.30, got {_STC1010_VECTOR[1]}"
        )


    def test_stc1010_vector_her2_zero():
        """her2 axis (index 4) must be 0.00."""
        try:
            from layer4_output.clinical_trials_routes import _STC1010_VECTOR
        except ImportError:
            pytest.skip("clinical_trials_routes not importable in test environment")

        assert abs(_STC1010_VECTOR[4] - 0.00) < 0.001
    ''')

print("\nW3 complete — all files written.")
````

````python

import os, textwrap

REPO = "/mnt/shared-workspace/shared/crispro-backend-v3-base"

def write(path, content):
    full = os.path.join(REPO, path)
    os.makedirs(os.path.dirname(full), exist_ok=True)
    with open(full, "w") as f:
        f.write(textwrap.dedent(content))
    print(f"  wrote {path} ({os.path.getsize(full)} bytes)")

# ── LAYER 1: atr_ddr_engine/__init__.py ──────────────────────────────────────
write("layer1_engines/atr_ddr_engine/__init__.py", '''
    """ATR/DDR Engine — Berzosertib, Adavosertib, CAPRI, LATIFY."""
    from .engine import AtrDdrEngine, evaluate_atr_ddr
    from .models import AtrDdrInput, AtrDdrResult, DDRProfile, TrialResult

    __all__ = [
        "AtrDdrEngine", "evaluate_atr_ddr",
        "AtrDdrInput", "AtrDdrResult", "DDRProfile", "TrialResult",
    ]
    ''')

# ── LAYER 1: atr_ddr_engine/constants.py ─────────────────────────────────────
write("layer1_engines/atr_ddr_engine/constants.py", '''
    """
    ATR/DDR Engine — All Exact Quantitative Values (KG v2 Locked)

    LATIFY delta is QUARANTINED — cite responder 0.8936 vs non-responder 0.6295 only.
    DDR 0.983 figure is QUARANTINED — DL-07 publication-blocking.

    RUO — Research Use Only.
    """

    # ── Berzosertib (ATR inhibitor) ───────────────────────────────────────────
    BERZOSERTIB_RS_HIGH_ORR = 0.40      # 40% — RS-High (replication stress high)
    BERZOSERTIB_RS_LOW_ORR = 0.05       # 5%  — RS-Low
    BERZOSERTIB_RS_HIGH_LABEL = "RS-High"
    BERZOSERTIB_RS_LOW_LABEL = "RS-Low"
    BERZOSERTIB_MECHANISM = "ATR inhibitor — blocks replication stress response"

    # ── Adavosertib (WEE1 inhibitor) ─────────────────────────────────────────
    ADAVOSERTIB_PTEN_INTACT_ORR = 0.23  # 23% — PTEN-intact
    ADAVOSERTIB_PTEN_LOSS_ORR = 0.00    # 0%  — PTEN-loss
    ADAVOSERTIB_PTEN_INTACT_PFS_HR = 0.55   # BENEFICIAL
    ADAVOSERTIB_PTEN_LOSS_PFS_HR = 1.82     # HARMFUL
    ADAVOSERTIB_MECHANISM = "WEE1 inhibitor — blocks G2/M checkpoint"
    ADAVOSERTIB_PTEN_LOSS_WARNING = (
        "PTEN-loss patients showed HARMFUL outcomes with adavosertib (WEE1 inhibitor): "
        "PFS HR=1.82. PTEN-loss is a contraindication for WEE1 inhibition."
    )

    # ── CAPRI trial ───────────────────────────────────────────────────────────
    CAPRI_PARPI_NAIVE_ORR = 0.36        # 36% — PARPi-naive
    CAPRI_POST_PARPI_ORR = 0.04         # 4%  — post-PARPi
    CAPRI_PARPI_NAIVE_PFS_HR = 0.42     # BENEFICIAL
    CAPRI_POST_PARPI_PFS_HR = 1.31      # HARMFUL
    CAPRI_MECHANISM = "ATR inhibitor + PARPi combination"
    CAPRI_POST_PARPI_WARNING = (
        "Post-PARPi patients showed HARMFUL outcomes in CAPRI: PFS HR=1.31. "
        "ATR inhibitor + PARPi combination is contraindicated post-PARPi."
    )

    # ── LATIFY trial ──────────────────────────────────────────────────────────
    # DELTA IS QUARANTINED — cite responder/non-responder values only
    LATIFY_RESPONDER_SCORE = 0.8936     # Responder score (cite this)
    LATIFY_NON_RESPONDER_SCORE = 0.6295 # Non-responder score (cite this)
    LATIFY_DELTA_STATUS = "QUARANTINED"
    LATIFY_DELTA_DISCLOSURE = (
        "LATIFY delta is QUARANTINED — vector version unresolved. "
        "Cite responder 0.8936 vs non-responder 0.6295 only. "
        "Do not cite the delta value."
    )
    LATIFY_MECHANISM = "ATR inhibitor (ceralasertib) + olaparib in BRCA-mutated ovarian cancer"

    # ── DDR 0.983 — QUARANTINED ───────────────────────────────────────────────
    DDR_0983_STATUS = "QUARANTINED"
    DDR_0983_DISCLOSURE = (
        "DDR 0.983 figure is QUARANTINED — DL-07 publication-blocking. "
        "Do not cite this value."
    )

    # ── RS score thresholds ───────────────────────────────────────────────────
    RS_HIGH_THRESHOLD = 0.5     # RS score ≥ 0.5 → RS-High
    RS_LOW_THRESHOLD = 0.5      # RS score < 0.5 → RS-Low

    # ── ATM mutation → ATR inhibitor eligibility ──────────────────────────────
    ATM_ATR_RATIONALE = (
        "ATM-mutated tumors are hypersensitive to ATR inhibition. "
        "ATM loss → replication fork instability → ATR dependency. "
        "RS-High ORR 40% vs RS-Low 5% (berzosertib)."
    )
    ''')

# ── LAYER 1: atr_ddr_engine/models.py ────────────────────────────────────────
write("layer1_engines/atr_ddr_engine/models.py", '''
    """ATR/DDR Engine — Data Models. RUO."""

    from dataclasses import dataclass, field
    from typing import List, Optional
    from enum import Enum


    class DDRProfile(str, Enum):
        RS_HIGH = "RS_HIGH"         # Replication stress high — ATR inhibitor eligible
        RS_LOW = "RS_LOW"           # Replication stress low
        PARPI_NAIVE = "PARPI_NAIVE" # PARPi-naive — CAPRI eligible
        POST_PARPI = "POST_PARPI"   # Post-PARPi — CAPRI contraindicated
        PTEN_INTACT = "PTEN_INTACT" # PTEN-intact — adavosertib eligible
        PTEN_LOSS = "PTEN_LOSS"     # PTEN-loss — adavosertib HARMFUL
        UNKNOWN = "UNKNOWN"


    @dataclass
    class TrialResult:
        trial: str
        agent: str
        subgroup: str
        orr: Optional[float]
        pfs_hr: Optional[float]
        interpretation: str  # "BENEFICIAL" / "HARMFUL" / "NEUTRAL"
        evidence_tier: str
        warning: Optional[str] = None


    @dataclass
    class LatifyResult:
        responder_score: float
        non_responder_score: float
        delta_status: str
        mandatory_disclosure: str


    @dataclass
    class AtrDdrInput:
        patient_id: str
        disease: str = "ovarian cancer"
        atm_status: Optional[str] = None    # "ATM_mut" / "WT"
        brca_status: Optional[str] = None   # "BRCA1_mut" / "BRCA2_mut" / "WT"
        pten_status: Optional[str] = None   # "PTEN_null" / "intact"
        rs_score: Optional[float] = None    # 0.0–1.0
        prior_parpi: bool = False


    @dataclass
    class AtrDdrResult:
        patient_id: str
        ddr_profile: DDRProfile
        trial_results: List[TrialResult]
        latify: LatifyResult
        recommended_agent: Optional[str]
        recommendation_rationale: str
        warnings: List[str]
        evidence_anchors: List[str]
        ruo: bool = True
        disclaimer: str = "Research Use Only. Not cleared for clinical decision-making."
    ''')

# ── LAYER 1: atr_ddr_engine/engine.py ────────────────────────────────────────
write("layer1_engines/atr_ddr_engine/engine.py", '''
    """
    ATR/DDR Engine

    Evaluates ATR/DDR therapy eligibility based on DDR profile.
    All values from KG v2 verified anchors.

    LATIFY delta: QUARANTINED — cite responder 0.8936 vs non-responder 0.6295 only.
    DDR 0.983: QUARANTINED — DL-07 publication-blocking.

    RUO — Research Use Only.
    """

    from typing import List, Optional
    from .models import AtrDdrInput, AtrDdrResult, DDRProfile, TrialResult, LatifyResult
    from .constants import (
        BERZOSERTIB_RS_HIGH_ORR, BERZOSERTIB_RS_LOW_ORR,
        ADAVOSERTIB_PTEN_INTACT_ORR, ADAVOSERTIB_PTEN_LOSS_ORR,
        ADAVOSERTIB_PTEN_INTACT_PFS_HR, ADAVOSERTIB_PTEN_LOSS_PFS_HR,
        ADAVOSERTIB_PTEN_LOSS_WARNING,
        CAPRI_PARPI_NAIVE_ORR, CAPRI_POST_PARPI_ORR,
        CAPRI_PARPI_NAIVE_PFS_HR, CAPRI_POST_PARPI_PFS_HR,
        CAPRI_POST_PARPI_WARNING,
        LATIFY_RESPONDER_SCORE, LATIFY_NON_RESPONDER_SCORE,
        LATIFY_DELTA_STATUS, LATIFY_DELTA_DISCLOSURE,
        RS_HIGH_THRESHOLD, ATM_ATR_RATIONALE,
    )


    class AtrDdrEngine:
        """Evaluates ATR/DDR therapy eligibility."""

        def evaluate(self, inp: AtrDdrInput) -> AtrDdrResult:
            profile = self._classify_profile(inp)
            trials = self._build_trial_results(inp, profile)
            warnings = self._collect_warnings(inp, profile)
            agent, rationale = self._recommend(inp, profile)

            return AtrDdrResult(
                patient_id=inp.patient_id,
                ddr_profile=profile,
                trial_results=trials,
                latify=LatifyResult(
                    responder_score=LATIFY_RESPONDER_SCORE,
                    non_responder_score=LATIFY_NON_RESPONDER_SCORE,
                    delta_status=LATIFY_DELTA_STATUS,
                    mandatory_disclosure=LATIFY_DELTA_DISCLOSURE,
                ),
                recommended_agent=agent,
                recommendation_rationale=rationale,
                warnings=warnings,
                evidence_anchors=[
                    f"Berzosertib: RS-High ORR {BERZOSERTIB_RS_HIGH_ORR:.0%} vs RS-Low {BERZOSERTIB_RS_LOW_ORR:.0%}",
                    f"Adavosertib: PTEN-intact ORR {ADAVOSERTIB_PTEN_INTACT_ORR:.0%} vs PTEN-loss {ADAVOSERTIB_PTEN_LOSS_ORR:.0%}",
                    f"Adavosertib: PTEN-intact PFS HR={ADAVOSERTIB_PTEN_INTACT_PFS_HR} (BENEFICIAL) vs PTEN-loss HR={ADAVOSERTIB_PTEN_LOSS_PFS_HR} (HARMFUL)",
                    f"CAPRI: PARPi-naive ORR {CAPRI_PARPI_NAIVE_ORR:.0%} (PFS HR={CAPRI_PARPI_NAIVE_PFS_HR}) vs post-PARPi {CAPRI_POST_PARPI_ORR:.0%} (PFS HR={CAPRI_POST_PARPI_PFS_HR})",
                    f"LATIFY: responder {LATIFY_RESPONDER_SCORE} vs non-responder {LATIFY_NON_RESPONDER_SCORE} — delta QUARANTINED",
                    "ATM_mut → ATR inhibitor eligible (ceralasertib/berzosertib)",
                ],
            )

        def _classify_profile(self, inp: AtrDdrInput) -> DDRProfile:
            if inp.rs_score is not None:
                return DDRProfile.RS_HIGH if inp.rs_score >= RS_HIGH_THRESHOLD else DDRProfile.RS_LOW
            if inp.pten_status == "PTEN_null":
                return DDRProfile.PTEN_LOSS
            if inp.pten_status == "intact":
                return DDRProfile.PTEN_INTACT
            if inp.prior_parpi:
                return DDRProfile.POST_PARPI
            if inp.atm_status == "ATM_mut":
                return DDRProfile.RS_HIGH  # ATM_mut → RS-High proxy
            return DDRProfile.UNKNOWN

        def _build_trial_results(self, inp: AtrDdrInput, profile: DDRProfile) -> List[TrialResult]:
            results = []
            # Berzosertib
            results.append(TrialResult(
                trial="Berzosertib (ATR inhibitor)",
                agent="berzosertib",
                subgroup="RS-High",
                orr=BERZOSERTIB_RS_HIGH_ORR,
                pfs_hr=None,
                interpretation="BENEFICIAL",
                evidence_tier="CLINICAL_STAGE",
            ))
            results.append(TrialResult(
                trial="Berzosertib (ATR inhibitor)",
                agent="berzosertib",
                subgroup="RS-Low",
                orr=BERZOSERTIB_RS_LOW_ORR,
                pfs_hr=None,
                interpretation="NEUTRAL",
                evidence_tier="CLINICAL_STAGE",
            ))
            # Adavosertib
            results.append(TrialResult(
                trial="Adavosertib (WEE1 inhibitor)",
                agent="adavosertib",
                subgroup="PTEN-intact",
                orr=ADAVOSERTIB_PTEN_INTACT_ORR,
                pfs_hr=ADAVOSERTIB_PTEN_INTACT_PFS_HR,
                interpretation="BENEFICIAL",
                evidence_tier="CLINICAL_STAGE",
            ))
            results.append(TrialResult(
                trial="Adavosertib (WEE1 inhibitor)",
                agent="adavosertib",
                subgroup="PTEN-loss",
                orr=ADAVOSERTIB_PTEN_LOSS_ORR,
                pfs_hr=ADAVOSERTIB_PTEN_LOSS_PFS_HR,
                interpretation="HARMFUL",
                evidence_tier="CLINICAL_STAGE",
                warning=ADAVOSERTIB_PTEN_LOSS_WARNING,
            ))
            # CAPRI
            results.append(TrialResult(
                trial="CAPRI",
                agent="ATR inhibitor + PARPi",
                subgroup="PARPi-naive",
                orr=CAPRI_PARPI_NAIVE_ORR,
                pfs_hr=CAPRI_PARPI_NAIVE_PFS_HR,
                interpretation="BENEFICIAL",
                evidence_tier="CLINICAL_STAGE",
            ))
            results.append(TrialResult(
                trial="CAPRI",
                agent="ATR inhibitor + PARPi",
                subgroup="post-PARPi",
                orr=CAPRI_POST_PARPI_ORR,
                pfs_hr=CAPRI_POST_PARPI_PFS_HR,
                interpretation="HARMFUL",
                evidence_tier="CLINICAL_STAGE",
                warning=CAPRI_POST_PARPI_WARNING,
            ))
            return results

        def _collect_warnings(self, inp: AtrDdrInput, profile: DDRProfile) -> List[str]:
            warnings = []
            if profile == DDRProfile.PTEN_LOSS:
                warnings.append(ADAVOSERTIB_PTEN_LOSS_WARNING)
            if inp.prior_parpi:
                warnings.append(CAPRI_POST_PARPI_WARNING)
            warnings.append(LATIFY_DELTA_DISCLOSURE)
            return warnings

        def _recommend(self, inp: AtrDdrInput, profile: DDRProfile):
            if profile == DDRProfile.RS_HIGH:
                return "berzosertib or ceralasertib", ATM_ATR_RATIONALE
            if profile == DDRProfile.PTEN_INTACT and not inp.prior_parpi:
                return "adavosertib", f"PTEN-intact ORR {ADAVOSERTIB_PTEN_INTACT_ORR:.0%}, PFS HR={ADAVOSERTIB_PTEN_INTACT_PFS_HR}"
            if profile == DDRProfile.PTEN_LOSS:
                return None, f"Adavosertib CONTRAINDICATED — PTEN-loss PFS HR={ADAVOSERTIB_PTEN_LOSS_PFS_HR} (HARMFUL)"
            if profile == DDRProfile.POST_PARPI:
                return None, f"CAPRI CONTRAINDICATED post-PARPi — PFS HR={CAPRI_POST_PARPI_PFS_HR} (HARMFUL)"
            return None, "Insufficient DDR profile data for recommendation"


    def evaluate_atr_ddr(inp: AtrDdrInput) -> AtrDdrResult:
        """Convenience function."""
        return AtrDdrEngine().evaluate(inp)
    ''')

# ── TESTS: atr_ddr_engine ─────────────────────────────────────────────────────
write("tests/test_layer1/test_atr_ddr_engine/__init__.py", "")
write("tests/test_layer1/test_atr_ddr_engine/test_engine.py", '''
    """Tests for ATR/DDR Engine — all exact values locked. RUO."""
    import pytest
    from layer1_engines.atr_ddr_engine import evaluate_atr_ddr, AtrDdrInput
    from layer1_engines.atr_ddr_engine.models import DDRProfile
    from layer1_engines.atr_ddr_engine.constants import (
        BERZOSERTIB_RS_HIGH_ORR, BERZOSERTIB_RS_LOW_ORR,
        ADAVOSERTIB_PTEN_INTACT_ORR, ADAVOSERTIB_PTEN_LOSS_ORR,
        ADAVOSERTIB_PTEN_INTACT_PFS_HR, ADAVOSERTIB_PTEN_LOSS_PFS_HR,
        CAPRI_PARPI_NAIVE_ORR, CAPRI_POST_PARPI_ORR,
        CAPRI_PARPI_NAIVE_PFS_HR, CAPRI_POST_PARPI_PFS_HR,
        LATIFY_RESPONDER_SCORE, LATIFY_NON_RESPONDER_SCORE,
        LATIFY_DELTA_STATUS, DDR_0983_STATUS,
    )


    def _inp(**kwargs):
        defaults = dict(patient_id="DDR-TEST-001", disease="ovarian cancer")
        defaults.update(kwargs)
        return AtrDdrInput(**defaults)


    # ── Berzosertib values ────────────────────────────────────────────────────
    def test_berzosertib_rs_high_orr():
        assert abs(BERZOSERTIB_RS_HIGH_ORR - 0.40) < 0.001

    def test_berzosertib_rs_low_orr():
        assert abs(BERZOSERTIB_RS_LOW_ORR - 0.05) < 0.001

    def test_berzosertib_rs_high_8x_rs_low():
        """RS-High ORR must be 8× RS-Low ORR."""
        assert BERZOSERTIB_RS_HIGH_ORR / BERZOSERTIB_RS_LOW_ORR == pytest.approx(8.0, abs=0.1)


    # ── Adavosertib values ────────────────────────────────────────────────────
    def test_adavosertib_pten_intact_orr():
        assert abs(ADAVOSERTIB_PTEN_INTACT_ORR - 0.23) < 0.001

    def test_adavosertib_pten_loss_orr_zero():
        assert abs(ADAVOSERTIB_PTEN_LOSS_ORR - 0.00) < 0.001

    def test_adavosertib_pten_intact_pfs_hr_beneficial():
        assert ADAVOSERTIB_PTEN_INTACT_PFS_HR < 1.0
        assert abs(ADAVOSERTIB_PTEN_INTACT_PFS_HR - 0.55) < 0.001

    def test_adavosertib_pten_loss_pfs_hr_harmful():
        assert ADAVOSERTIB_PTEN_LOSS_PFS_HR > 1.0
        assert abs(ADAVOSERTIB_PTEN_LOSS_PFS_HR - 1.82) < 0.001


    # ── CAPRI values ──────────────────────────────────────────────────────────
    def test_capri_parpi_naive_orr():
        assert abs(CAPRI_PARPI_NAIVE_ORR - 0.36) < 0.001

    def test_capri_post_parpi_orr():
        assert abs(CAPRI_POST_PARPI_ORR - 0.04) < 0.001

    def test_capri_parpi_naive_pfs_hr_beneficial():
        assert CAPRI_PARPI_NAIVE_PFS_HR < 1.0
        assert abs(CAPRI_PARPI_NAIVE_PFS_HR - 0.42) < 0.001

    def test_capri_post_parpi_pfs_hr_harmful():
        assert CAPRI_POST_PARPI_PFS_HR > 1.0
        assert abs(CAPRI_POST_PARPI_PFS_HR - 1.31) < 0.001


    # ── LATIFY values — QUARANTINED ───────────────────────────────────────────
    def test_latify_responder_score():
        assert abs(LATIFY_RESPONDER_SCORE - 0.8936) < 0.0001

    def test_latify_non_responder_score():
        assert abs(LATIFY_NON_RESPONDER_SCORE - 0.6295) < 0.0001

    def test_latify_delta_quarantined():
        assert LATIFY_DELTA_STATUS == "QUARANTINED"

    def test_ddr_0983_quarantined():
        assert DDR_0983_STATUS == "QUARANTINED"


    # ── Engine integration tests ──────────────────────────────────────────────
    def test_atm_mut_rs_high_profile():
        result = evaluate_atr_ddr(_inp(atm_status="ATM_mut"))
        assert result.ddr_profile == DDRProfile.RS_HIGH

    def test_atm_mut_recommends_atr_inhibitor():
        result = evaluate_atr_ddr(_inp(atm_status="ATM_mut"))
        assert result.recommended_agent is not None
        assert "berzosertib" in result.recommended_agent or "ceralasertib" in result.recommended_agent

    def test_pten_loss_harmful_no_recommendation():
        result = evaluate_atr_ddr(_inp(pten_status="PTEN_null"))
        assert result.recommended_agent is None
        assert "HARMFUL" in result.recommendation_rationale

    def test_pten_intact_recommends_adavosertib():
        result = evaluate_atr_ddr(_inp(pten_status="intact"))
        assert result.recommended_agent == "adavosertib"

    def test_post_parpi_no_recommendation():
        result = evaluate_atr_ddr(_inp(prior_parpi=True))
        assert result.recommended_agent is None
        assert "HARMFUL" in result.recommendation_rationale

    def test_latify_always_quarantined_in_result():
        result = evaluate_atr_ddr(_inp())
        assert result.latify.delta_status == "QUARANTINED"
        assert "QUARANTINED" in result.latify.mandatory_disclosure

    def test_latify_responder_scores_in_result():
        result = evaluate_atr_ddr(_inp())
        assert abs(result.latify.responder_score - 0.8936) < 0.0001
        assert abs(result.latify.non_responder_score - 0.6295) < 0.0001

    def test_six_trial_results_always_present():
        result = evaluate_atr_ddr(_inp())
        assert len(result.trial_results) == 6

    def test_pten_loss_warning_in_warnings():
        result = evaluate_atr_ddr(_inp(pten_status="PTEN_null"))
        assert any("1.82" in w for w in result.warnings)

    def test_ruo_flag():
        result = evaluate_atr_ddr(_inp())
        assert result.ruo is True

    def test_evidence_anchors_complete():
        result = evaluate_atr_ddr(_inp())
        anchors = " ".join(result.evidence_anchors)
        assert "40%" in anchors
        assert "5%" in anchors
        assert "23%" in anchors
        assert "1.82" in anchors
        assert "0.42" in anchors
        assert "1.31" in anchors
        assert "0.8936" in anchors
        assert "QUARANTINED" in anchors
    ''')

print("\nW3 complete — ATR/DDR engine + 30 tests written.")
````

````python

# W3: Clerk auth integration - write all auth files for frontend v3

import os

DST = "/workspace/crispro-frontend-v3"

# ── src/context/ClerkAuthContext.jsx ─────────────────────────────────────────
clerk_auth_ctx = '''/**
 * ClerkAuthContext — replaces Supabase AuthContext for CrisPRO v3.
 * Wraps Clerk hooks and exposes the same interface as the old AuthContext
 * so existing components that use useAuth() work without changes.
 *
 * Persona is read from Clerk user.publicMetadata.persona
 * Partner ID is read from Clerk user.publicMetadata.partner_id
 */
import React, { createContext, useContext, useMemo } from 'react';
import { useUser, useAuth as useClerkAuth, useClerk } from '@clerk/clerk-react';

const AuthContext = createContext({});

export const useAuth = () => {
  const ctx = useContext(AuthContext);
  if (!ctx) throw new Error('useAuth must be used within AuthProvider');
  return ctx;
};

export const AuthProvider = ({ children }) => {
  const { user, isLoaded: userLoaded } = useUser();
  const { isSignedIn, getToken } = useClerkAuth();
  const { signOut } = useClerk();

  const persona = user?.publicMetadata?.persona || 'pharma';
  const partnerId = user?.publicMetadata?.partner_id || 'brenus';

  const value = useMemo(() => ({
    // Legacy interface compatibility
    user: user ? {
      id: user.id,
      email: user.primaryEmailAddress?.emailAddress,
      user_metadata: { full_name: user.fullName },
    } : null,
    session: isSignedIn ? { user } : null,
    authenticated: !!isSignedIn,
    loading: !userLoaded,
    profile: user ? {
      user_id: user.id,
      email: user.primaryEmailAddress?.emailAddress,
      full_name: user.fullName,
      persona,
      partner_id: partnerId,
      tier: user.publicMetadata?.tier || 'partner',
    } : null,
    profileLoading: !userLoaded,
    // Clerk-specific
    persona,
    partnerId,
    getToken,
    signOut: () => signOut(),
    // Stub for legacy Supabase calls (no-ops)
    signIn: async () => ({ error: 'Use Clerk SignIn component' }),
    signUp: async () => ({ error: 'Use Clerk SignIn component' }),
  }), [user, isSignedIn, userLoaded, persona, partnerId, getToken, signOut]);

  return <AuthContext.Provider value={value}>{children}</AuthContext.Provider>;
};

export default AuthProvider;
'''

# ── src/context/PersonaContext.jsx (updated with pharma persona) ──────────────
persona_ctx = '''/**
 * PersonaContext — v3 persona access matrix.
 * Personas: pharma (active), patient (scaffolded), hospital (scaffolded)
 */
import React, { createContext, useContext, useMemo } from 'react';
import { useAuth } from './ClerkAuthContext';

const PersonaContext = createContext({});

export const usePersona = () => {
  const ctx = useContext(PersonaContext);
  if (!ctx) throw new Error('usePersona must be used within PersonaProvider');
  return ctx;
};

const PERSONA_ACCESS = {
  pharma: {
    pages: [
      '/pharma',
      '/pharma/dashboard',
      '/pharma/rag',
      '/pharma/fit-gap',
      '/pharma/sl',
      '/pharma/trials',
      '/pharma/dossier',
      '/pharma/metastasis',
      '/pharma/ddr',
      '/pharma/orchestrator',
      '/pharma/research',
    ],
    features: [
      'rag_chat',
      'fit_gap_analysis',
      'sl_engine',
      'trial_intelligence',
      'dossier_browser',
      'governance_status',
      'metastasis_dashboard',
      'ddr_status',
      'orchestrator',
    ],
    defaultRoute: '/pharma/dashboard',
    label: 'Pharma Partner',
  },
  patient: {
    pages: ['/patient'],
    features: [],
    defaultRoute: '/patient',
    label: 'Patient',
    comingSoon: true,
  },
  hospital: {
    pages: ['/hospital'],
    features: [],
    defaultRoute: '/hospital',
    label: 'Hospital',
    comingSoon: true,
  },
};

export const PersonaProvider = ({ children }) => {
  const { persona, authenticated } = useAuth();

  const value = useMemo(() => {
    const access = PERSONA_ACCESS[persona] || PERSONA_ACCESS.pharma;
    return {
      persona,
      access,
      canAccess: (path) => {
        if (!authenticated) return false;
        return access.pages.some(p => path.startsWith(p));
      },
      hasFeature: (feature) => access.features.includes(feature),
      defaultRoute: access.defaultRoute,
      isPharma: persona === 'pharma',
      isPatient: persona === 'patient',
      isHospital: persona === 'hospital',
      isComingSoon: access.comingSoon || false,
    };
  }, [persona, authenticated]);

  return <PersonaContext.Provider value={value}>{children}</PersonaContext.Provider>;
};

export default PersonaProvider;
'''

# ── src/components/auth/ClerkSignIn.jsx ──────────────────────────────────────
clerk_signin = '''/**
 * ClerkSignIn — custom sign-in page with no Clerk branding.
 * Uses Clerk's <SignIn> component with appearance overrides.
 */
import React from 'react';
import { SignIn } from '@clerk/clerk-react';
import { Box, Typography } from '@mui/material';

const clerkAppearance = {
  elements: {
    // Remove Clerk footer ("Secured by Clerk")
    footer: { display: 'none' },
    footerAction: { display: 'none' },
    footerActionLink: { display: 'none' },
    // Remove Clerk logo
    logoBox: { display: 'none' },
    logoImage: { display: 'none' },
    // Style the card to match CrisPRO dark theme
    card: {
      backgroundColor: '#1e293b',
      border: '1px solid #334155',
      borderRadius: '12px',
      boxShadow: '0 25px 50px -12px rgba(0,0,0,0.5)',
    },
    headerTitle: { color: '#f1f5f9', fontFamily: 'inherit' },
    headerSubtitle: { color: '#94a3b8', fontFamily: 'inherit' },
    formFieldLabel: { color: '#cbd5e1', fontFamily: 'inherit' },
    formFieldInput: {
      backgroundColor: '#0f172a',
      border: '1px solid #334155',
      color: '#f1f5f9',
      fontFamily: 'inherit',
    },
    formButtonPrimary: {
      backgroundColor: '#0891b2',
      fontFamily: 'inherit',
      '&:hover': { backgroundColor: '#0e7490' },
    },
    identityPreviewText: { color: '#94a3b8' },
    dividerLine: { backgroundColor: '#334155' },
    dividerText: { color: '#64748b' },
    socialButtonsBlockButton: {
      backgroundColor: '#1e293b',
      border: '1px solid #334155',
      color: '#f1f5f9',
    },
  },
  variables: {
    fontFamily: 'inherit',
    colorPrimary: '#06b6d4',
    colorBackground: '#1e293b',
    colorText: '#f1f5f9',
    colorTextSecondary: '#94a3b8',
    colorInputBackground: '#0f172a',
    colorInputText: '#f1f5f9',
    borderRadius: '8px',
  },
};

export default function ClerkSignIn() {
  return (
    <Box
      sx={{
        minHeight: '100vh',
        display: 'flex',
        flexDirection: 'column',
        alignItems: 'center',
        justifyContent: 'center',
        bgcolor: '#0f172a',
        px: 2,
      }}
    >
      {/* CrisPRO branding — no Clerk mention */}
      <Box sx={{ mb: 4, textAlign: 'center' }}>
        <Typography
          variant="h4"
          sx={{ color: '#06b6d4', fontWeight: 900, letterSpacing: '-0.02em', mb: 0.5 }}
        >
          CrisPRO
        </Typography>
        <Typography variant="body2" sx={{ color: '#64748b' }}>
          Precision Oncology Intelligence Platform
        </Typography>
      </Box>

      <SignIn
        appearance={clerkAppearance}
        redirectUrl="/pharma/dashboard"
        signUpUrl={false}  // disable sign-up link
      />

      <Typography variant="caption" sx={{ color: '#334155', mt: 3 }}>
        RUO — Research Use Only. Not cleared for clinical decision-making.
      </Typography>
    </Box>
  );
}
'''

# ── src/components/auth/PersonaRoute.jsx (updated for Clerk) ─────────────────
persona_route = '''/**
 * PersonaRoute — route guard using Clerk auth + persona access matrix.
 * Replaces the old Supabase-based PersonaRoute.
 */
import React from 'react';
import { Navigate } from 'react-router-dom';
import { useAuth } from '../../context/ClerkAuthContext';
import { usePersona } from '../../context/PersonaContext';
import { Box, Typography, CircularProgress } from '@mui/material';

export default function PersonaRoute({ allowedPersonas, children }) {
  const { authenticated, loading } = useAuth();
  const { persona, isComingSoon } = usePersona();

  if (loading) {
    return (
      <Box sx={{ display: 'flex', justifyContent: 'center', alignItems: 'center', minHeight: '60vh' }}>
        <CircularProgress sx={{ color: '#06b6d4' }} />
      </Box>
    );
  }

  if (!authenticated) {
    return <Navigate to="/sign-in" replace />;
  }

  if (allowedPersonas && !allowedPersonas.includes(persona)) {
    if (isComingSoon) {
      return <ComingSoonPage persona={persona} />;
    }
    return <Navigate to="/pharma/dashboard" replace />;
  }

  return children;
}

function ComingSoonPage({ persona }) {
  return (
    <Box
      sx={{
        minHeight: '80vh',
        display: 'flex',
        flexDirection: 'column',
        alignItems: 'center',
        justifyContent: 'center',
        bgcolor: '#0f172a',
        color: '#f1f5f9',
        textAlign: 'center',
        px: 4,
      }}
    >
      <Typography variant="h3" sx={{ fontWeight: 900, color: '#06b6d4', mb: 2 }}>
        Coming Soon
      </Typography>
      <Typography variant="h6" sx={{ color: '#94a3b8', mb: 1 }}>
        The {persona} portal is under development.
      </Typography>
      <Typography variant="body2" sx={{ color: '#64748b', maxWidth: 400 }}>
        CrisPRO v3 is currently optimized for pharma partners. Patient and hospital
        portals are on the roadmap. Contact the CrisPRO team to be notified.
      </Typography>
    </Box>
  );
}
'''

# ── src/lib/apiConfig.js (updated for v3) ────────────────────────────────────
api_config = '''/**
 * API Configuration — CrisPRO v3
 * Supports v2 (legacy) and v3 (production) backends via env var toggle.
 */

const API_VERSION = import.meta.env.VITE_API_VERSION || 'v3';

const API_ROOTS = {
  v2: import.meta.env.VITE_API_ROOT_V2 || 'https://crispro-backend-v2.onrender.com',
  v3: import.meta.env.VITE_API_ROOT_V3 || import.meta.env.VITE_API_ROOT || 'http://localhost:8000',
};

export const API_ROOT = API_ROOTS[API_VERSION] || API_ROOTS.v3;
export const RAG_ROOT = `${API_ROOT}/rag`;
export const PHARMA_ROOT = `${API_ROOT}/pharma`;
export const API_V1_ROOT = `${API_ROOT}/api/v1`;

export const IS_V3 = API_VERSION === 'v3';
export const IS_V2 = API_VERSION === 'v2';

export default API_ROOT;
'''

# ── src/hooks/useRAGChat.js ───────────────────────────────────────────────────
use_rag_chat = '''/**
 * useRAGChat — React hook for Brenus RAG chat.
 * Manages session, message history, and API calls.
 */
import { useState, useCallback, useRef } from 'react';
import { useAuth } from '../context/ClerkAuthContext';
import { RAG_ROOT } from '../lib/apiConfig';

export function useRAGChat({ partnerId = 'brenus', autoEvaluate = true } = {}) {
  const { getToken } = useAuth();
  const [messages, setMessages] = useState([]);
  const [loading, setLoading] = useState(false);
  const [error, setError] = useState(null);
  const sessionIdRef = useRef(null);

  const sendMessage = useCallback(async (query) => {
    if (!query.trim()) return;
    setLoading(true);
    setError(null);

    // Optimistic user message
    const userMsg = { role: 'user', content: query, timestamp: new Date().toISOString() };
    setMessages(prev => [...prev, userMsg]);

    try {
      const token = await getToken();
      const resp = await fetch(`${RAG_ROOT}/chat/${partnerId}`, {
        method: 'POST',
        headers: {
          'Content-Type': 'application/json',
          ...(token ? { Authorization: `Bearer ${token}` } : {}),
        },
        body: JSON.stringify({
          query,
          session_id: sessionIdRef.current,
          partner_id: partnerId,
          top_k: 5,
          evaluate: autoEvaluate,
        }),
      });

      if (!resp.ok) {
        const err = await resp.json().catch(() => ({}));
        throw new Error(err.detail || `HTTP ${resp.status}`);
      }

      const data = await resp.json();
      sessionIdRef.current = data.session_id;

      const assistantMsg = {
        role: 'assistant',
        content: data.answer,
        citations: data.citations || [],
        faithfulness: data.faithfulness,
        answer_relevancy: data.answer_relevancy,
        composite_score: data.composite_score,
        passed: data.passed,
        crispro_cta: data.crispro_cta,
        guardrail_violations: data.guardrail_violations || [],
        timestamp: new Date().toISOString(),
      };
      setMessages(prev => [...prev, assistantMsg]);
      return assistantMsg;
    } catch (err) {
      setError(err.message);
      setMessages(prev => prev.slice(0, -1)); // remove optimistic message
    } finally {
      setLoading(false);
    }
  }, [getToken, partnerId, autoEvaluate]);

  const clearSession = useCallback(() => {
    sessionIdRef.current = null;
    setMessages([]);
    setError(null);
  }, []);

  return {
    messages,
    loading,
    error,
    sessionId: sessionIdRef.current,
    sendMessage,
    clearSession,
  };
}
'''

files = {
    f"{DST}/src/context/ClerkAuthContext.jsx": clerk_auth_ctx,
    f"{DST}/src/context/PersonaContext.jsx": persona_ctx,
    f"{DST}/src/components/auth/ClerkSignIn.jsx": clerk_signin,
    f"{DST}/src/components/auth/PersonaRoute.jsx": persona_route,
    f"{DST}/src/lib/apiConfig.js": api_config,
    f"{DST}/src/hooks/useRAGChat.js": use_rag_chat,
}

for path, content in files.items():
    os.makedirs(os.path.dirname(path), exist_ok=True)
    with open(path, "w") as f:
        f.write(content)
    print(f"✓ {path.replace(DST+'/', '')}")

print("\nW3 Clerk auth files written.")
````
